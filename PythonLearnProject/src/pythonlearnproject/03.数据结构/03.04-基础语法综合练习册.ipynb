## 一、算术运算：`//` 和 `%` 是一对搭档


````python
a = 100
b = 3
print('100 // 3 =', a // b)     # 33  整除，向下取整
print('100 %  3 =', a % b)      # 1   余数
print('验证：3 * 33 + 1 =', b * (a // b) + a % b)   # 正好等于 100

# 【实用】用 divmod 一次拿到商和余
print('divmod(100, 3) =', divmod(a, b))

# 【实用】秒数转时分秒
seconds = 7385
print(f'{seconds} 秒 = {seconds // 3600} 时 {seconds % 3600 // 60} 分 {seconds % 60} 秒')
````

## 二、列表：最基础的增删查改


````python
thislist = ['apple', 'banana', 'cherry', 'orange', 'kiwi', 'melon', 'mango']

print('负索引 [-1]     ->', thislist[-1])        # 最后一个
print('切片 [2:5]      ->', thislist[2:5])       # 下标 2,3,4，不含 5
print('切片 [::2]      ->', thislist[::2])       # 隔一个取一个
print('反转 [::-1]     ->', thislist[::-1])      # 反着来
print('长度 len()      ->', len(thislist))

# 增
lst = ['apple', 'banana', 'cherry']
lst.append('orange')            # 末尾追加
print('append          ->', lst)
lst.insert(1, 'pear')           # 在指定位置插入
print('insert(1, ...)   ->', lst)

# 删
lst.remove('pear')              # 按值删
print('remove          ->', lst)
lst.pop(0)                      # 按位置删
print('pop(0)          ->', lst)

# 拼接：两种方式的区别
list1 = ['a', 'b', 'c']
list2 = [1, 2, 3]

list3 = list1 + list2           # 方式 1：+ 号拼接，生成【新列表】
print('list1 + list2    ->', list3)
print('原 list1 没变    ->', list1)

list4 = ['a', 'b', 'c']
for x in list2:
    list4.append(x)             # 方式 2：逐个追加，改的是【原列表】
print('逐个 append      ->', list4)

# 【区别】想要"元素级相加"（1+1=2 那种），上面两种都不对，要用 zip：
nums_a = [1, 2, 3]
nums_b = [10, 20, 30]
print('元素级相加       ->', [x + y for x, y in zip(nums_a, nums_b)])
````

## 三、元组：只能读，不能改


````python
thistuple = ('apple', 'banana', 'cherry', 'orange', 'kiwi', 'melon', 'mango')

print('切片 [-4:-1]  ->', thistuple[-4:-1])      # ('orange', 'kiwi', 'melon')
print('长度          ->', len(thistuple))
print('成员判断      ->', 'apple' in thistuple)

# 遍历
for item in thistuple[:3]:                       # 只遍历前三个
    print('  ', item)

# 想改内容？列表和元组互转
t = ('apple', 'banana', 'cherry')
as_list = list(t)
as_list[1] = 'kiwi'                              # 在列表阶段修改
t = tuple(as_list)
print('绕一圈改完    ->', t)

# 【实用】元组解包，一行拿到多个值
point = (31.2, 121.5)
lat, lng = point
print(f'纬度 {lat}，经度 {lng}')
````

## 四、集合：去重和"有没有"


````python
thisset = {'apple', 'banana', 'cherry'}
print('"banana" in thisset  ->', 'banana' in thisset)
print('"grape" in thisset   ->', 'grape' in thisset)

# 集合最实用的场景：去重
duplicated = [1, 2, 2, 3, 3, 3, 'a', 'a']
print('去重前：', duplicated)
print('去重后：', list(set(duplicated)))
print('保序去重（更常用）：', list(dict.fromkeys(duplicated)))
# 注意：set 不保证顺序，所以输出顺序可能是乱的；
# dict.fromkeys 从 Python 3.7 起保证按插入顺序，是"保序去重"的标准写法。

# 集合运算
a, b = {1, 2, 3, 4}, {3, 4, 5, 6}
print('交集（都有）    ->', a & b)
print('并集（合起来）  ->', a | b)
print('差集（a 有 b 没有）->', a - b)
print('对称差（只在一方）->', a ^ b)
````

## 五、字典：按名字找东西


````python
thisdict = {
    'brand': 'Ford',
    'model': 'Mustang',
    'year': 1964,
}
print('原始字典：', thisdict)

# 取值的两种方式
print('thisdict["model"]     ->', thisdict['model'])          # 键不存在会 KeyError
print('thisdict.get("price") ->', thisdict.get('price'))      # 键不存在返回 None（更安全）
print('带默认值            ->', thisdict.get('price', '暂无报价'))

# 【重要】用 get 避免 KeyError
for key in ['brand', 'price', 'year']:
    print(f'  {key}: {thisdict.get(key, "（没有这个字段）")}')

# 增 / 改 / 删
thisdict['color'] = 'red'          # 键不存在就是新增
thisdict['year'] = 1965            # 键存在就是修改
print('增改之后：', thisdict)

del thisdict['color']
print('删除之后：', thisdict)

# 遍历字典的三种姿势
print('\n遍历键：', end=' ')
for k in thisdict:
    print(k, end=' ')

print('\n遍历值：', end=' ')
for v in thisdict.values():
    print(v, end=' ')

print('\n遍历键值对：')
for k, v in thisdict.items():
    print(f'  {k} = {v}')

# 【实用】把两个列表合成字典
keys = ['语文', '数学', '英语']
values = [90, 85, 78]
print('zip 合成字典 ->', dict(zip(keys, values)))

# 嵌套字典：真实项目里最常见的结构
students = {
    '学生1': {'姓名': '张三', '年龄': '20', '地址': '武汉'},
    '学生2': {'姓名': '李四', '年龄': '21', '地址': '北京'},
}
for sid, info in students.items():
    print(f'{sid}：{info["姓名"]}，{info["年龄"]} 岁，来自 {info["地址"]}')
````

## 七、格式化输出与类型转换


````python
c = 10.45
result3 = '数字是：%d' % c
print('用 %d 处理浮点数 ->', result3, '（注意：截断了，不是四舍五入）')
print('正确的保留小数  ->', '数字是：%.2f' % c)

# f-string 的三种常用格式
print(f'保留两位小数：{c:.2f}')
print(f'百分比：{0.8567:.1%}')
print(f'千分位：{1234567:,}')

# 类型转换：字符串转数字再运算
print(int('10') + 5, float('3.14') * 2, str(100) + 'abc')

# 从字符串里"安全地"取数字
raw = '  42 '
print('去掉空白再转：', int(raw.strip()))
````

## 八、条件判断：分支与嵌套分支


````python
# 场景：游乐园买票 —— 身高 + VIP 等级双重判断（嵌套 if）
def ticket(has_height, vip_level):
    if has_height:
        print('  您身高已达标')
        if vip_level > 3:
            print('  尊敬的 VIP 用户，您可以免费游玩')
        else:
            print(f'  VIP 等级只有 {vip_level}，需要付费 10 元')
    else:
        print('  欢迎免费游玩（身高未达标，免票）')


print('身高达标 + VIP 5：')
ticket(True, 5)
print('身高达标 + VIP 2：')
ticket(True, 2)
print('身高不达标：')
ticket(False, 1)
````

**嵌套 if 的读法**：先过第一道门槛，只在通过时才继续看第二道。
上面的逻辑等价于下面这种"扁平化"写法，可读性更好：


````python
def ticket_flat(has_height, vip_level):
    if not has_height:
        return '欢迎免费游玩（身高未达标，免票）'
    if vip_level > 3:
        return '尊敬的 VIP 用户，您可以免费游玩'
    return f'VIP 等级只有 {vip_level}，需要付费 10 元'


print(ticket_flat(True, 5))
print(ticket_flat(True, 2))
print(ticket_flat(False, 1))
# 用"提前返回"代替嵌套，是让代码变清爽的重要技巧。
````

## 九、循环统计：计数、求和、条件筛选


````python
# 统计字符串里 'a' 出现的次数
name = 'itheima is a brand of itcast'
count = 0
for ch in name:
    if ch == 'a':
        count += 1
print(f'name 中共有 {count} 个 a')

# 更简洁的写法
print('用 count()  ->', name.count('a'))
print('用 sum()    ->', sum(1 for ch in name if ch == 'a'))

# 统计 1~99 里偶数的个数
evens = 0
for x in range(1, 100):
    if x % 2 == 0:
        evens += 1
print('1~99 里偶数个数：', evens)
print('公式验证：99 // 2 =', 99 // 2)

# 生成偶数列表
print('偶数列表：', [x for x in range(1, 100) if x % 2 == 0][:10], '...')

# 【重要认知】循环变量的作用域
x = 1000
for x in range(1, 4):        # for 的循环变量会"覆盖"外面的同名变量
    print('循环内 x =', x)
print('循环结束后 x =', x)    # 3 —— Python 的 for 变量会泄漏到外层，这点和 C/Java 不同
````

## 十、经典题目：九九乘法表


````python
# 写法 1：外层 for + 内层 for
for i in range(1, 10):
    for j in range(1, i + 1):
        print(f'{i}*{j} = {i * j}\t', end='')
    print()
````

````python
# 写法 2：用 join 一次拼出一行（更"Python"，但对初学者稍难）
for i in range(1, 10):
    row = '\t'.join(f'{i}*{j} = {i * j}' for j in range(1, i + 1))
    print(row)
````

**两个写法导出的图形不一样，注意 `i*j` 的顺序**：

- 写法 1 的列数随行号增加（1 列、2 列、3 列……）
- 如果交换 `i` 和 `j`，得到的就是镜像的图形


````python
# 交换顺序看看差异
for i in range(1, 10):
    for j in range(1, i + 1):
        print(f'{j}*{i} = {i * j}\t', end='')
    print()
````

## 十一、综合案例：绩效发工资

**需求**：公司有 20 个员工，每人随机一个 1~10 的绩效分。

- 绩效分 < 5：不满足条件，下一位
- 绩效分 >= 5 且公司账户余额 >= 1000：发 1000 元，账户扣减
- 账户余额不足：停止发钱

这个例子把**循环 + 分支 + `continue` + `break` + 累减**都用上了，
是很好的综合练习。


````python
money = 10000          # 公司账户余额
paid = 0               # 已发人数
random.seed(7)         # 固定种子，方便复现结果

for i in range(1, 21):
    score = random.randint(1, 10)

    if score < 5:
        print(f'员工{i}的绩效分 {score}，不满足，下一位')
        continue                       # 跳过本轮剩下的代码，直接下一位

    if money >= 1000:
        money -= 1000
        paid += 1
        print(f'员工{i}的绩效分 {score}，满足，发工资 1000，账户余额 {money}')
    else:
        print(f'账户余额不足，当前余额 {money}，停止发放')
        break                          # 直接跳出整个循环

print(f'\n总共发了 {paid} 人的工资，账户剩余 {money} 元')
````

**`continue` 和 `break` 的区别在这里体现得最清楚**：

- `continue`：这一位不发，但**下一位还要继续看** → 用于"跳过"
- `break`：钱不够了，**后面所有员工都不用看了** → 用于"终止"

用错这两个关键字，程序逻辑会完全不同。判断技巧：
**问自己"这个情况发生后，剩下的人还要不要处理？"** 要 → `continue`，不要 → `break`。

## 十三、常见坑清单

| 坑 | 现象 | 正解 |
|---|---|---|
| 用 `list1 + list2` 想元素相加 | 得到拼接后的长列表 | `[x+y for x,y in zip(a,b)]` |
| `d['不存在的键']` | `KeyError` | `d.get('键', 默认值)` |
| 用 `set()` 去重后顺序乱 | 输出顺序不可控 | `list(dict.fromkeys(x))` 保序去重 |
| `while` 忘了推进变量 | 死循环 | 循环体里必须有 `i += 1` 之类 |
| `continue` 当成 `break` 用 | 该停的没停 | 想清楚"剩下还要不要处理" |
| `for` 的循环变量名复用了外层变量 | 外层变量被覆盖 | 换个名字，或用函数隔离 |
| `%d` 处理浮点数 | 小数被截断 | 用 `%.2f` 或 f-string |
| 推断了 `input()` 的返回类型 | 字符串和数字比较报错 | 先 `int()` / `float()` |
| 用嵌套 if 套太多层 | 代码难读 | 用"提前 return"扁平化 |
