这份笔记是整个项目的**总目录和复习地图**。
目录结构已经按 **`01` ~ `10` 编号**排好，编号就是**建议的学习顺序**，
每个模块内部的文件也按 `01.01`、`01.02` 这样排，**照着数字一路往下走就行**。

### 命名规则（一眼看懂目录）

| 形式 | 含义 | 例子 |
|---|---|---|
| `NN.名称/` | **模块**（第 NN 个学习阶段） | `07.pandas/` |
| `NN.MM-名称.ipynb` | 模块内**第 MM 篇**笔记 | `07.05-DataFrame.ipynb` |
| 不带编号的文件 | **数据/产物**，不是学习内容 | `data.txt`、`生成文件/` |

> **为什么编号要补零？** 因为不补零的话，字符串排序会把 `10` 排到 `2` 前面，
> 文件夹里的顺序就乱了。写成 `01`、`02` … `10` 才能永远按顺序排。


````python
# ---- 先看看整个项目的目录长什么样 ----
import os
from pathlib import Path

# 找出项目根目录（本笔记所在的那一层）
def find_root():
    here = Path.cwd()
    for base in (here, here / 'src' / 'pythonlearnproject', here.parent):
        if (base / '00.学习路线.ipynb').exists():
            return base
    return here


ROOT = find_root()
print('项目根目录：', ROOT)
print()

for d in sorted(p for p in ROOT.iterdir() if p.is_dir()):
    nbs = sorted(f.name for f in d.glob('*.ipynb'))
    print(f'{d.name}/   （{len(nbs)} 篇笔记）')
````

## 四、自动目录（永远不会过期）

下面这段代码会**实时扫描**目录，列出所有笔记。
以后你新增/改名了笔记，**不用改这份索引**，重新运行这一格就行。


````python
def print_tree(root, indent=''):
    """递归打印目录树，标出每个模块的篇数。"""
    dirs = sorted(p for p in root.iterdir() if p.is_dir()
                  and not p.name.startswith(('.', '_')))
    nbs = sorted(f.name for f in root.glob('*.ipynb'))

    for nb in nbs:
        print(f'{indent}    {nb}')

    for d in dirs:
        count = len(list(d.glob('*.ipynb')))
        extra = f'（{count} 篇）' if count else '（数据/产物）'
        print(f'{indent}[DIR] {d.name} {extra}')
        print_tree(d, indent + '    ')


print('=' * 60)
print('项目完整目录树')
print('=' * 60)
print(f'{ROOT.name}/')
print_tree(ROOT)

total = len(list(ROOT.rglob('*.ipynb')))
print()
print('=' * 60)
print(f'合计：{total} 篇 Notebook')
````

### 一个必须知道的结构约束

模块目录名带 `01.` 这种编号前缀，**好处是排序清晰**，
但也带来一个副作用：**带数字开头和点号的目录，不能作为 Python 包的导入路径**
（`import 05.模块与包` 是语法错误）。

所以：

| 目录 | 能不能 import |
|---|---|
| `01.基础概念/` ~ `10.MySQL/` | ❌ 只能当文件夹 |
| `05.模块与包/my_uyils/`（里面那层） | ✅ 可以 import |

具体怎么处理，看 `05.01-模块与包.ipynb`，那里讲得很细。
