# 一、pandas 是什么

**pandas 是一个"带标签的表格处理库"**。它的价值在于：
现实世界的数据大多是**表格**（Excel、数据库表、CSV），
而 Python 原生的 `list` / `dict` 处理表格非常笨拙。

pandas 建立在 **numpy** 之上，
可以理解成「**numpy 负责算得快，pandas 负责看得懂**」：

| | numpy | pandas |
|---|---|---|
| 核心对象 | `ndarray`（纯数值数组） | `Series` / `DataFrame` |
| 有列名吗 | ❌ 只有位置 | ✅ 有列名、有行索引 |
| 能混类型吗 | ❌ 一种 dtype | ✅ 每列可以不同 |
| 适合 | 数值计算、矩阵运算 | **表格数据分析** |

pandas 的五个核心能力：

| 能力 | 说明 |
|---|---|
| **标签化索引** | 用"姓名""日期"这样的标签定位数据，而不是记下标 |
| **缺失值处理** | 内置 `NaN`，`isna` / `fillna` / `dropna` 一整套工具 |
| **自动对齐** | 两个表相加时自动按标签对齐，不用手动对位置 |
| **强大的 IO** | 一行代码读写 CSV / Excel / SQL / JSON / Parquet |
| **时间序列** | 日期范围生成、频率转换、重采样（resample） |

# 三、为什么要用 pandas：一次直接对比

先别急着学 API。看看**同一个需求，手写和 pandas 差多远**，
你就知道这些 API 值不值得学了。

需求：从一份员工数据里，算出**每个职位的平均薪资**。


````python
# 用最"朴素"的 Python 写法
employees = [
    {'job': '软件工程师', 'salary': 20000},
    {'job': '产品经理', 'salary': 25000},
    {'job': '软件工程师', 'salary': 18000},
    {'job': '客服代表', 'salary': 8000},
    {'job': '产品经理', 'salary': 28000},
    {'job': '软件工程师', 'salary': 22000},
]

# ① 先分组求和 + 计数
acc = {}
for e in employees:
    job = e['job']
    if job not in acc:
        acc[job] = [0, 0]          # [总和, 个数]
    acc[job][0] += e['salary']
    acc[job][1] += 1

# ② 再算平均
manual = {job: total / count for job, (total, count) in acc.items()}
print('手写循环的结果：', manual)
print('代码行数：约 12 行')
````

````python
# pandas 的写法
df_demo = pd.DataFrame(employees)
print('pandas 的结果：')
result = df_demo.groupby('job', as_index=False)['salary'].mean()
print(result)
print('代码行数：2 行')
````

> **差距不只是"行数少"**。更关键的是：
>
> - 手写版要自己维护"累加器"和"计数器"，逻辑一复杂就容易写错
> - 手写版要自己处理**缺失值、异常值、空分组**
> - pandas 版想换成"求中位数"，只要把 `mean()` 改成 `median()`
>
> 顺带说一个 Python 自带的小工具 —— **`set` 集合**，
> 它天然去重，也能做"快速看有几种取值"这种小事：


````python
# set 集合：无序、自动去重
print('set 去重：', set([1, 2, 3, 3, 3, 2]))

jobs_set = {e['job'] for e in employees}
print('有哪些职位：', jobs_set)
print('职位数量：', len(jobs_set))

print('\n同样的信息，pandas 用 nunique()：')
print('有哪些职位：', df_demo['job'].unique().tolist())
print('职位数量：', df_demo['job'].nunique())
````

# 六、数据读写：四种常见格式

先说结论表，再逐个演示：

| 格式 | 读 | 写 | 要点 |
|---|---|---|---|
| CSV | `read_csv` | `to_csv` | 中文用 `encoding='utf-8-sig'` |
| Excel | `read_excel` | `to_excel` | 要装 `openpyxl`；加 `index=False` |
| TXT | `read_csv(sep='\t')` | `to_csv(sep='\t')` | **TXT 本质就是换了分隔符的 CSV** |
| JSON | `read_json` | `to_json` | 中文要 `force_ascii=False` |


````python
OUT_DIR = os.path.join(os.getcwd(), '生成文件', 'pandas01')
os.makedirs(OUT_DIR, exist_ok=True)
print('演示文件输出目录：', OUT_DIR)
````

## 6.1 CSV（最常用）


````python
csv_path = os.path.join(OUT_DIR, 'employee_data.csv')
df.to_csv(csv_path, index=False, encoding='utf-8-sig')
print('已写入：', csv_path)

back = pd.read_csv(csv_path, encoding='utf-8-sig')
print('读回来形状：', back.shape)
print('和原表一样吗：', back.shape == df.shape)
````

> **`encoding='utf-8-sig'` 为什么必须加？**
>
> `utf-8-sig` 比普通 `utf-8` 多了 3 个字节的 **BOM**（字节序标记）。
> Excel 靠这个标记识别"这是 UTF-8"，
> 不加的话，用 Excel 打开中文就变乱码。
>
> 顺手验证一下 BOM 真的存在：


````python
print('前 6 个字节：', open(csv_path, 'rb').read(6))
print('开头是不是 BOM(EF BB BF)：', open(csv_path, 'rb').read(3) == b'\xef\xbb\xbf')
````

## 6.2 Excel


````python
xlsx_path = os.path.join(OUT_DIR, 'employee_data.xlsx')
df.to_excel(xlsx_path, index=False, engine='openpyxl')
print('已写入：', xlsx_path)

back_x = pd.read_excel(xlsx_path)
print('读回来形状：', back_x.shape)

# 只读前 3 行、只读指定列，能省内存
print('\n只读部分数据（nrows=3, usecols=["employ_id","job_name"]）：')
print(pd.read_excel(xlsx_path, nrows=3, usecols=['employ_id', 'job_name']))
````

## 6.3 TXT

TXT 没有统一的"格式标准"，所以在 pandas 眼里
**它就是"用某个分隔符分开的文本"** —— 所以还是用 `read_csv`，只是换 `sep`。


````python
txt_path = os.path.join(OUT_DIR, 'employee_data.txt')
df.to_csv(txt_path, sep='\t', index=False, encoding='utf-8-sig')     # 制表符分隔
print('已写入（制表符分隔）：', txt_path)

print('\n文件原本长这样：')
with open(txt_path, encoding='utf-8-sig') as f:
    for _ in range(3):
        print(repr(f.readline()))

print('\n读回来：')
back_t = pd.read_csv(txt_path, sep='\t')
print(back_t.head(3))
print('形状：', back_t.shape)
````

## 6.4 JSON


````python
json_path = os.path.join(OUT_DIR, 'employee_data.json')

# orient='records' -> [{"列": 值}, ...] 这种"一行一个对象"的结构，最适合给外部系统用
df.head(5).to_json(json_path, orient='records', force_ascii=False, indent=2)

print('已写入，文件开头：')
print(open(json_path, encoding='utf-8').read()[:260], '...')

print('\n读回来：')
back_j = pd.read_json(json_path, orient='records')
print(back_j)
print('形状：', back_j.shape)
````

> **`force_ascii=False` 为什么重要？**
>
> 不加这个参数，中文会被转义成 `\u51e4\u82f1` 这种"乱码"，人能看懂才有鬼。
> 加了它，文件里才是真正的中文。
>
> 另外还要注意**日期**：默认会写成"毫秒时间戳"
> （像 `1704067200000`），完全没法阅读，而且新版 pandas 已经把这个默认值标记为废弃 ——
> 运行下面的单元格时会看到一条 `Pandas4Warning`，正是它在提醒你。
> 建议总是显式写 `date_format='iso'`：


````python
# 日期在 JSON 里怎么存？默认居然是一串"毫秒时间戳"，很不直观
date_demo = pd.DataFrame({'日期': pd.to_datetime(['2024-01-01', '2024-01-02']), '值': [1, 2]})

p_default = os.path.join(OUT_DIR, 'date_default.json')
date_demo.to_json(p_default, orient='records', force_ascii=False)
print('默认写出来长这样：')
print(' ', open(p_default, encoding='utf-8').read())
print('  -> 1704067200000 是"距离 1970-01-01 的毫秒数"，人根本看不懂')
````

````python
# 【正解】显式用 date_format='iso'，存成人能看懂的日期字符串
p_iso = os.path.join(OUT_DIR, 'date_iso.json')
date_demo.to_json(p_iso, orient='records', force_ascii=False, date_format='iso')
print("用 date_format='iso' 写出来：")
print(' ', open(p_iso, encoding='utf-8').read())

print('\n读回来（日期是字符串，要自己转）：')
back = pd.read_json(p_iso, orient='records')
print(back)
print('原始类型：', back.dtypes.to_dict())
print('转成日期后：', pd.to_datetime(back['日期']).dtype)
````

# 七、一个很典型的坑：电话号码被读成了数字

看看前面 `df.head()` 的 `phone_number` 列，或者 `df.dtypes` ——
它被 pandas 认成了 **数字**，而不是字符串。

这看起来无害，但**电话号码、身份证号、邮编、订单号**这类"用数字写的编号"，
当数字处理会出各种问题：


````python
print('phone_number 当前的 dtype：', df['phone_number'].dtype)
print('前 3 个值：', df['phone_number'].head(3).tolist())
print('-> 它被当成了数字。好在原始手机号是 11 位、没有前导 0，所以暂时没出事。')

# 但换成"带前导 0 的编号"（比如座机、工号、身份证号），问题立刻暴露：
demo = pd.DataFrame({'工号': ['0012345', '0012346', '0100088']})   # 纯数字，但有前导 0
print('\n原始（字符串，前导 0 还在）：')
print(demo)
print('dtype：', demo['工号'].dtype)

demo.to_csv(os.path.join(OUT_DIR, 'staff_id.csv'), index=False, encoding='utf-8-sig')
reloaded = pd.read_csv(os.path.join(OUT_DIR, 'staff_id.csv'), encoding='utf-8-sig')
print('\n存成 CSV 再读回来：')
print(reloaded)
print('dtype：', reloaded['工号'].dtype)
print('-> 前导 0 全没了！0012345 变成了 12345，数据被【静默篡改】了。')
````

````python
# 【正解】读的时候显式声明 dtype
fixed = pd.read_csv(
    os.path.join(OUT_DIR, 'staff_id.csv'),
    encoding='utf-8-sig',
    dtype={'工号': str},        # 告诉 pandas：这列是字符串，别猜
)
print('显式指定 dtype 之后：')
print(fixed)
print('dtype：', fixed['工号'].dtype)
print('-> 前导 0 保住了 ✅')
````

> **一句话记住**：**"用数字组成的编号"不是数字。**
> 电话号码、身份证号、邮编、银行卡号、订单号，全部应该当**字符串**存。
> 读文件时用 `dtype={列名: str}` 显式指定，别让 pandas 猜。

# 常见坑清单

| 坑 | 现象 | 正解 |
|---|---|---|
| CSV 中文乱码 | Excel 打开是问号 | 写/读都加 `encoding='utf-8-sig'` |
| 电话号码变数字 | 前导 0 丢失、变成科学计数法 | 读时 `dtype={'电话': str}` |
| `to_excel` 多了行号列 | 表里多一列 0,1,2... | 加 `index=False` |
| `read_json` 报错 | 传入的是 JSON 字符串却被当路径 | 用 `io.StringIO(...)` 包起来 |
| 日期读回来是字符串 | 不能做时间运算 | `pd.to_datetime()` 转 |
| 只读大文件全部进内存 | 内存爆炸 | `nrows=` / `usecols=` 限制 |
| 路径写死成 `C:/Users/xxx/...` | 换电脑就崩 | 用相对路径或变量 |
| 忘记了 `sheet_name` | 只读到第一个 sheet | `sheet_name=None` 读全部 |
