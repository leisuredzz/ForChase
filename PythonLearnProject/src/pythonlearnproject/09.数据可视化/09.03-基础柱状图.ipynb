# 柱状图：比较"谁多谁少"

---

**学习目标**

1. 掌握 `Bar` 的基本三步走
2. 会做横向条形图（`reversal_axis`）—— 名字长的时候更好看
3. 会画多系列对比图和堆叠柱状图
4. 会给柱子加数值标签

**一句话理解**

柱状图回答的问题是"**哪个多、哪个少**"。
只要数据是"几个项目各自的总量"，就该用它。

**选择指南**

| 想表达什么 | 用什么 |
|---|---|
| 各项数值对比 | **柱状图**（纵向） |
| 项目名字很长 | **条形图**（横向，`reversal_axis`） |
| 各项占比 | 饼图 / 环形图 |
| 随时间变化 | 折线图 |


````python
from pyecharts.charts import Bar
from pyecharts import options as opts
````

---
## 一、基本三步走：创建 → 加数据 → 出 HTML


````python
bar = Bar()

# x 轴：分类名称
bar.add_xaxis(['衬衫', '羊毛衫', '雪纺衫', '裤子', '高跟鞋', '袜子'])

# y 轴：对应数值
bar.add_yaxis('商家A', [5, 20, 36, 10, 75, 90])

bar.set_global_opts(
    title_opts=opts.TitleOpts(title='各商品销量'),
    yaxis_opts=opts.AxisOpts(name='销量（件）'),
)

bar.render('基础柱状图.html')
print('已生成 基础柱状图.html')
````

---
## 二、横向条形图：名字太长时的救星

调用 `reversal_axis()` 就把 x/y 轴互换，柱子变成横着躺。

**为什么需要它？** 纵向柱状图里，分类名是**竖着斜着**显示在 x 轴上的，
名字一长就挤成一团。横向条形图的分类名可以**水平排开**，清爽得多。


````python
bar_h = Bar()
bar_h.add_xaxis(['衬衫', '羊毛衫', '雪纺衫', '裤子', '高跟鞋', '袜子'])
bar_h.add_yaxis('商家A', [5, 20, 36, 10, 75, 90])

# 这一行就是全部改动
bar_h.reversal_axis()

# 【注意】反转坐标轴后，数值标签的位置也要跟着改：'right'（右）而不是 'top'（上）
bar_h.set_global_opts(
    title_opts=opts.TitleOpts(title='各商品销量（横向）'),
    xaxis_opts=opts.AxisOpts(name='销量（件）'),
)
bar_h.render('横向条形图.html')
print('已生成 横向条形图.html')
````

**`reversal_axis()` 的一个小坑**：调用它之后，
原本"y 轴"的配置项现在对应到的是屏幕上的 x 轴，
所以 `xaxis_opts` / `yaxis_opts` 的语义也要跟着调过来。
很多人第一次用会发现自己设置的轴名称跑到了另一条轴上，就是这个原因。


---
## 三、多系列对比：同一种商品，不同商家的销量

想在同一张图里比较多个系列，就调用多次 `add_yaxis`。


````python
bar_multi = Bar()
bar_multi.add_xaxis(['衬衫', '羊毛衫', '雪纺衫', '裤子', '高跟鞋', '袜子'])

bar_multi.add_yaxis('商家A', [5, 20, 36, 10, 75, 90])
bar_multi.add_yaxis('商家B', [15, 6, 45, 20, 35, 66])
bar_multi.add_yaxis('商家C', [25, 32, 18, 40, 55, 30])

bar_multi.set_global_opts(
    title_opts=opts.TitleOpts(title='三家商家销量对比'),
    legend_opts=opts.LegendOpts(pos_top='8%'),
    xaxis_opts=opts.AxisOpts(name='商品'),
    yaxis_opts=opts.AxisOpts(name='销量（件）'),
    tooltip_opts=opts.TooltipOpts(trigger='axis', axis_pointer_type='shadow'),
)
bar_multi.render('多系列柱状图.html')
print('已生成 多系列柱状图.html')
````

---
## 四、堆叠柱状图：既看总量，又看构成

给同一组 `add_yaxis` 传**相同的 `stack` 值**，它们就会堆在一起。

**什么时候用堆叠？** 当你想同时表达"总量"和"各部分占比"时。
而不关心总量、只想比较各系列时，就用上面的并排柱状图。


````python
bar_stack = Bar()
bar_stack.add_xaxis(['1月', '2月', '3月', '4月', '5月'])

# 注意 stack 参数：值相同的柱子会叠起来
bar_stack.add_yaxis('线上', [120, 132, 101, 134, 90], stack='总量')
bar_stack.add_yaxis('线下', [220, 182, 191, 234, 290], stack='总量')
bar_stack.add_yaxis('批发', [150, 232, 201, 154, 190], stack='总量')

bar_stack.set_global_opts(
    title_opts=opts.TitleOpts(title='各渠道销售额构成（堆叠）'),
    legend_opts=opts.LegendOpts(pos_top='8%'),
    yaxis_opts=opts.AxisOpts(name='销售额（万元）'),
)
bar_stack.render('堆叠柱状图.html')
print('已生成 堆叠柱状图.html')
````

---
## 五、数值标签：把数字直接写在柱子上

pyecharts 默认**不显示**柱子上的数值（鼠标悬停才看得到）。
想直接显示，用 `label_opts`：


````python
bar_label = Bar()
bar_label.add_xaxis(['衬衫', '羊毛衫', '雪纺衫', '裤子', '高跟鞋', '袜子'])
bar_label.add_yaxis(
    '商家A',
    [5, 20, 36, 10, 75, 90],
    label_opts=opts.LabelOpts(
        is_show=True,       # 显示标签
        position='top',     # 位置：top / inside / right / left
        font_size=12,
        color='#333',
    ),
)
bar_label.set_global_opts(
    title_opts=opts.TitleOpts(title='带数值标签的柱状图'),
    yaxis_opts=opts.AxisOpts(name='销量（件）'),
)
bar_label.render('带标签柱状图.html')
print('已生成 带标签柱状图.html')
````

> **注意**：柱子很多、或者数值相差很大时，标签会重叠，反而更难读。
> 这种情况下建议**关掉标签**，靠悬停来看具体数值。


---
## 六、横向 + 标签的组合（实战最常用的形态）

把"横向条形图"和"数值标签"结合起来，就是最常见的排行榜样式。


````python
rank_data = [
    ('Python', 92),
    ('JavaScript', 78),
    ('Java', 71),
    ('C#', 62),
    ('C++', 58),
    ('Go', 45),
]

bar_rank = Bar()
# 按数值升序排一下，横向图里这样能让最大的排在顶部
rank_data.sort(key=lambda x: x[1])
bar_rank.add_xaxis([name for name, _ in rank_data])
bar_rank.add_yaxis(
    '热度',
    [score for _, score in rank_data],
    label_opts=opts.LabelOpts(is_show=True, position='right'),
)
bar_rank.reversal_axis()
bar_rank.set_global_opts(
    title_opts=opts.TitleOpts(title='编程语言热度排行', pos_left='center'),
    xaxis_opts=opts.AxisOpts(name='热度指数'),
    legend_opts=opts.LegendOpts(is_show=False),
    tooltip_opts=opts.TooltipOpts(trigger='axis', axis_pointer_type='shadow'),
)
bar_rank.render('排行榜样式.html')
print('已生成 排行榜样式.html')
# 注意 add_xaxis 里用的是列表推导式：[name for name, _ in rank_data]
# 这是"从元组列表里只取第一列"的常见写法。
````

---
## 七、常见坑清单

| 坑 | 现象 | 正解 |
|---|---|---|
| `add_yaxis` 名字重复 | 图例里分不清 | 每个系列用不同的名字 |
| 用堆叠却忘了统一 `stack` 值 | 柱子并排而不是叠起来 | `stack` 值必须完全一致 |
| `reversal_axis()` 后轴名称错位 | 名字跑到另一条轴上 | 反向理解 `xaxis_opts` 的位置 |
| 分类名太长 | 纵向图里斜着挤在一起 | 改用横向（`reversal_axis`） |
| 数值标签太密 | 文字重叠看不清 | `is_show=False`，靠悬停查看 |
| x 轴数据顺序乱 | 排行榜看起来没规律 | 先 `sort(key=...)` 再传进去 |
| 忘了 `render()` | 没有 HTML 文件 | 最后一步一定要调用 |
| 柱子上不显示数字 | pyecharts 默认不显示 | `label_opts=opts.LabelOpts(is_show=True)` |
| 用 `add_yaxis` 传了字符串数字 | 图表空白或报错 | 数值要保持数字类型 |
