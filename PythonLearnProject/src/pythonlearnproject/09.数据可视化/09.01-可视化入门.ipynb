**一句话理解**

pyecharts 是"Python 版的 ECharts"。
它不画图片，而是**生成一个 HTML 文件**，用浏览器来看 ——
所以你得到的图表可以**缩放、悬停、拖动**，这是它和 matplotlib 最大的区别。


---
## 一、pyecharts 和 matplotlib 该选哪个？

| | matplotlib | pyecharts |
|---|---|---|
| 输出形式 | **静态图片**（png） | **交互式网页**（html） |
| 交互能力 | 无 | 悬停显示数值、缩放、图例开关 |
| 适合场景 | 论文插图、批量生成图片、数据分析探索 | 汇报大屏、网页嵌入、地图 |
| 安装 | `uv add matplotlib` | `uv add pyecharts` |

**简单判断**：要贴进文档 → matplotlib；要给人看、要"炫" → pyecharts。

> 本项目已经通过 `uv add pyecharts` 装好了依赖。
> 如果提示 `ModuleNotFoundError: No module named 'pyecharts'`，
> 在项目根目录执行 `uv add pyecharts` 即可。


````python
from pyecharts.charts import Line
from pyecharts import options as opts

# 先看一眼版本，出错时方便排查
import pyecharts
print('pyecharts 版本：', pyecharts.__version__)
````

## 二、第一个折线图：三步走

画任何 pyecharts 图表都是这个套路：

```python
1. 创建图表对象       chart = Line()
2. 添加数据           chart.add_xaxis(...) / chart.add_yaxis(...)
3. 生成 HTML          chart.render('文件名.html')
```

记住这个"三步走"，后面所有图表类型（Bar / Pie / Map / Scatter）都是同一套流程。


````python
line = Line()

# x 轴：国家名
line.add_xaxis(['China', 'English', 'American'])

# y 轴：GDP 数据。第一个参数是"这条线叫什么"，会显示在图例里
line.add_yaxis('GDP', [30, 20, 10])

# 生成 HTML 文件（默认文件名是 render.html）
line.render('入门折线图.html')
print('已生成 入门折线图.html，用浏览器打开看看')
````

**打开生成的 HTML 的方式**：

- 在 VS Code 的资源管理器里右键文件 → **Open with Live Server**，或
- 直接在文件管理器里双击（会用默认浏览器打开）

你会看到一条可以悬停、可以点击图例隐藏的折线 —— 这就是"交互式"的意思。

## 三、全局配置项：让图表好看又信息完整

`set_global_opts()` 负责设置**整个图表**的公共属性。
常用的四个：

| 配置项 | 作用 | 典型参数 |
|---|---|---|
| `TitleOpts` | 标题 | `title`、`pos_left`、`pos_bottom` |
| `LegendOpts` | 图例 | `is_show` |
| `ToolboxOpts` | 右上角工具箱（下载图片等） | `is_show` |
| `VisualMapOpts` | 视觉映射（颜色深浅表示数值大小） | `is_show`、`is_piecewise` |

> **重要提示**：新版 pyecharts 里这些配置类的导入路径有两个，
> 两种写法都能用：
>
> ```python
> from pyecharts.options import TitleOpts   # 推荐
> from pyecharts import options as opts     # 然后写 opts.TitleOpts(...)
> ```
>
> 原始笔记用的是第一种，本笔记统一用 `opts.XXX` 的形式，**更不容易和同名变量冲突**。


````python
line2 = Line()
line2.add_xaxis(['China', 'English', 'American'])
line2.add_yaxis('GDP', [30, 20, 10])

line2.set_global_opts(
    # 标题：放在底部居中
    title_opts=opts.TitleOpts(
        title='GDP 展示',
        pos_left='center',      # 水平位置：center / left / right / 百分比
        pos_bottom='1%',        # 垂直位置：从底部算 1%
    ),
    # 图例：默认开着，显式写出来更清楚
    legend_opts=opts.LegendOpts(is_show=True),
    # 工具箱：右上角的"下载图片 / 缩放 / 还原"按钮
    toolbox_opts=opts.ToolboxOpts(is_show=True),
    # 视觉映射：把数值大小映射成颜色深浅
    visualmap_opts=opts.VisualMapOpts(is_show=True),
)

line2.render('全局配置项示例.html')
print('已生成 全局配置项示例.html')
````

---
## 四、`render()` 的三种用法

| 写法 | 效果 |
|---|---|
| `chart.render()` | 生成默认文件 `render.html` |
| `chart.render('名字.html')` | 生成指定文件名的 HTML |
| `chart.render_notebook()` | **在 notebook 里直接渲染**（不用生成文件） |

> 注意：`render_notebook()` 依赖 notebook 环境能执行 JS，
> 如果显示不出来，**用 `render()` 生成 HTML 再打开最稳**。


````python
line3 = Line()
line3.add_xaxis(['一月', '二月', '三月', '四月'])
line3.add_yaxis('销售额', [120, 200, 150, 80])
line3.set_global_opts(title_opts=opts.TitleOpts(title='一行代码改标题'))

print(line3.render_embed()[:200], '...')     # render_embed 返回 HTML 字符串，便于嵌入网页

line3.render('季度销售额.html')
````

## 五、更多配置项：让图表更专业

`opts` 里还有不少好用的配置，这里挑最常用的几个试试。


````python
line4 = Line()
line4.add_xaxis(['周一', '周二', '周三', '周四', '周五'])
line4.add_yaxis(
    '访问量',
    [820, 932, 901, 934, 1290],
    # 平滑曲线（去掉折线的"尖角"）
    is_smooth=True,
    # 显示每个点的数值标签
    label_opts=opts.LabelOpts(is_show=True, position='top'),
    # 标记点、标记线
    markpoint_opts=opts.MarkPointOpts(
        data=[
            opts.MarkPointItem(type_='max', name='最大值'),
            opts.MarkPointItem(type_='min', name='最小值'),
        ]
    ),
    markline_opts=opts.MarkLineOpts(
        data=[opts.MarkLineItem(type_='average', name='平均值')]
    ),
)

line4.set_global_opts(
    title_opts=opts.TitleOpts(title='一周访问量（带标记）'),
    xaxis_opts=opts.AxisOpts(name='星期'),
    yaxis_opts=opts.AxisOpts(name='访问次数'),
    # 提示框：鼠标悬停时显示的内容
    tooltip_opts=opts.TooltipOpts(trigger='axis', axis_pointer_type='cross'),
    # 数据缩放：图表下方出现可拖动的缩放条
    datazoom_opts=opts.DataZoomOpts(),
)

line4.render('访问量示例.html')
print('已生成 访问量示例.html')
````

## 六、多系列对比：add_yaxis 调用多次

注意：**`add_xaxis` 只需要调用一次**（x 轴是共用的），
想画几条线就调用几次 `add_yaxis`。


````python
line5 = Line()
line5.add_xaxis(['1月', '2月', '3月', '4月', '5月'])
line5.add_yaxis('线上', [120, 132, 101, 134, 90])
line5.add_yaxis('线下', [220, 182, 191, 234, 290])
line5.add_yaxis('批发', [150, 232, 201, 154, 190])
line5.set_global_opts(
    title_opts=opts.TitleOpts(title='三种渠道销售额对比'),
    legend_opts=opts.LegendOpts(pos_top='5%'),      # 图例挪到上方，避免遮住标题
)
line5.render('多渠道对比.html')
print('已生成 多渠道对比.html')
````

## 七、主题：一行代码换皮肤

pyecharts 内置了多套主题，在创建图表时传 `init_opts=opts.InitOpts(theme=...)` 即可。


````python
from pyecharts.globals import ThemeType

print('可用主题（部分）：')
print([t for t in dir(ThemeType) if not t.startswith('_')][:20])

line6 = Line(init_opts=opts.InitOpts(theme=ThemeType.LIGHT, width='800px', height='400px'))
line6.add_xaxis(['A', 'B', 'C', 'D'])
line6.add_yaxis('数据', [10, 20, 30, 40])
line6.set_global_opts(title_opts=opts.TitleOpts(title='浅色主题'))
line6.render('主题示例.html')
print('已生成 主题示例.html')
# 常用主题：LIGHT / DARK / CHALK / ESSOS / INFOGRAPHIC / MACARONS / ROMA ...
````

## 八、常见坑清单

| 坑 | 现象 | 正解 |
|---|---|---|
| `ModuleNotFoundError: pyecharts` | 导入失败 | `uv add pyecharts` |
| `add_yaxis` 忘了写系列名 | 图例里是空白 / 报错 | 第一个参数必传图例名 |
| 反复调用 `add_xaxis` | x 轴数据被后一次覆盖 | x 轴只加一次 |
| 以为 `render()` 会弹窗 | 什么都没发生 | 它是**生成 HTML 文件**，要自己去打开 |
| HTML 打开是空白 | 数据没加进去 / 文件名写错 | 检查 add_xaxis 是否有数据 |
| `LabelOpts` 让图很乱 | 点太密，标签重叠 | 数据点多时 `is_show=False` |
| 中文标题乱码 | 极少见于 pyecharts | 确保 HTML 用 UTF-8 打开 |
| 图例挡住标题 | 位置默认在顶部 | `legend_opts=opts.LegendOpts(pos_top='5%')` |
| 用 `import pyecharts` 就想画图 | 提示没有 Line | 要 `from pyecharts.charts import Line` |
