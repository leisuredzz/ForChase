# Matplotlib 之四：饼图（pie）

---

饼图用来展示**各部分占整体的百分比**——比如预算分配、市场份额、时间都花在哪了。整张饼是 100%，每一块是一个比例。

## 学习目标

1. 用 `plt.pie()` 画饼图，理解 `labels` / `autopct` / `startangle` / `colors`
2. 搞清楚 `autopct='%1.1f%%'` 里 `%%` 到底是什么意思（原稿这里写错了）
3. 会用 `explode` 突出某一块、`shadow` 加阴影
4. 画出"环形图"（中间挖空的饼图）
5. 知道饼图**不适合**什么场景


---
## 一、什么时候用饼图？

| 适合 | 不适合 |
|---|---|
| 成分占比、部分/整体关系 | 类别太多（>6 块就难分辨） |
| 强调"谁最大" | 要精确比较数值大小（改用柱状图更准） |
| 比例相加正好 100% | 数据不是比例关系 |

> **关键认知**：人眼判断"扇区面积/角度"很不精确。如果只是比大小，**柱状图往往比饼图更清楚**。饼图胜在"一眼看出占比"。


---
## 二、最基础的饼图

`plt.pie(x, labels=, autopct=, startangle=, colors=)`：

- `x`：每块的数值（matplotlib 自动按比例算角度）
- `labels`：每块旁边的文字
- `autopct`：**扇区里显示的百分比文字**
- `startangle`：从几点钟方向开始画第一块（90 = 正上方）
- `colors`：每块颜色

> **原稿问题（重要）**：原稿注释写 `autopct='%1.1f%%'` 时说是"前后都添加 %，表示显示百分比"——这是**错的**。`%%` 是 Python 老式格式字符串里的**转义写法，表示"字面一个百分号 %"**（类似 `print("%d%%" % 50)` 输出 `50%`）。`%1.1f` 才是"保留 1 位小数的浮点数"。所以整句意思是：**用保留 1 位小数的浮点数显示，后面跟一个字面 % 号**。


````python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

subjects = ['Math', 'Physics', 'Chemistry', 'Biology', 'History']
scores = [80, 90, 75, 85, 95]
colors = ['#FF9999', '#66B2FF', '#99FF99', '#FFCC99', '#FF99FF']

# autopct='%1.1f%%'  ->  %1.1f 是"1位小数浮点数"，%% 是"字面的百分号"
plt.pie(scores,
        labels=subjects,
        autopct='%1.1f%%',
        startangle=90,    # 从正上方(12点)开始，饼图更"正"
        colors=colors)

plt.title('Student Scores')
plt.show()   # 输出: 5 色扇区，每块标着百分比
````

**角度小抄**：`startangle` 以"右侧 3 点钟"为 0°，逆时针增加：90°=正上(12点)、180°=正左(9点)、270°=正下(6点)。设 90° 让第一块从正上方开始，视觉最舒服。


---
## 三、进阶：突出某块 + 阴影

- `explode`：每块"炸开"的偏移量（元组，和 `x` 等长），`0` 不炸、越大越往外
- `shadow=True`：加阴影更有立体感
- `pctdistance`：百分比文字离圆心的距离（默认 0.6）


````python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

subjects = ['Math', 'Physics', 'Chemistry', 'Biology', 'History']
scores = [80, 90, 75, 85, 95]
colors = ['#FF9999', '#66B2FF', '#99FF99', '#FFCC99', '#FF99FF']
explode = (0.1, 0, 0, 0, 0)   # 只把第一块(Math)炸出来突出

plt.pie(scores,
        labels=subjects,
        autopct='%1.1f%%',
        startangle=90,
        colors=colors,
        explode=explode,
        shadow=True,
        pctdistance=0.7)

plt.title('Student Scores（突出 Math）')
plt.show()   # 输出: Math 那块被往外拽出一点，带阴影
````

## 四、环形图（中间挖空）

环形图 = 饼图 + `wedgeprops={'width': 0.5}`（扇区变"环带"，中间留出空洞），常在中空处写总结文字。

`wedgeprops={'width': 0.5}` 表示环带占半径的一半；`width` 越小环越细。再用 `plt.text(0, 0, ...)` 在圆心写字。


````python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

subjects = ['Math', 'Physics', 'Chemistry', 'Biology', 'History']
scores = [80, 90, 75, 85, 95]
colors = ['#FF9999', '#66B2FF', '#99FF99', '#FFCC99', '#AD99FF']
explode = (0.1, 0, 0, 0, 0)

plt.pie(scores,
        labels=subjects,
        autopct='%1.1f%%',
        startangle=90,
        colors=colors,
        pctdistance=0.7,
        explode=explode,
        wedgeprops={'width': 0.5})   # 【关键】width<1 变环形

# 在正中 (0,0) 写总结文字
plt.text(0, 0, 'Total\nScore\n100%',
         ha='center', va='center', fontsize=16, fontweight='bold')

plt.title('Student Scores（环形图）')
plt.show()   # 输出: 中间镂空的环形饼图，圆心写着 Total Score 100%
````

## 六、常见坑清单

| 坑 | 现象 | 正解 |
|---|---|---|
| `autopct` 里 `%` 报错/显示怪 | 不知道 `%%` 是转义 | `%%`=字面百分号；`%1.1f`=1位小数 |
| 扇区顺序"不正" | 第一块从右边开始 | `startangle=90` 从正上方起 |
| 类别太多 | 看不清谁大谁小 | >6 块考虑改用柱状图 |
| 想精确比大小 | 饼图不擅长 | 改用 `bar` |
| 中文方块 | labels 变 □ | 设 `font.sans-serif` 中文字体 |

## 七、动手练习

**练习 1**：画一个饼图，数据为 `[30, 25, 20, 15, 10]`，类别 `['A','B','C','D','E']`，显示百分比。

**练习 2**：给练习 1 的 C 块加 `explode=0.15` 突出，并 `shadow=True`。

**练习 3**：把练习 1 改成环形图（加 `wedgeprops={'width': 0.4}`），圆心写"占比"。

**练习 4**：用中文类别 `['吃饭','睡觉','学习','娱乐']` 和对应小时数画饼图，确认中文正常。

---

**答案思路（先自己做）**

```python
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

data = [30, 25, 20, 15, 10]; labels = ['A','B','C','D','E']

# 1
plt.figure(); plt.pie(data, labels=labels, autopct='%1.1f%%'); plt.show()

# 2
plt.figure()
plt.pie(data, labels=labels, autopct='%1.1f%%',
        explode=(0,0,0.15,0,0), shadow=True); plt.show()

# 3 环形
plt.figure()
plt.pie(data, labels=labels, autopct='%1.1f%%',
        wedgeprops={'width': 0.4})
plt.text(0, 0, '占比', ha='center', va='center'); plt.show()
```
