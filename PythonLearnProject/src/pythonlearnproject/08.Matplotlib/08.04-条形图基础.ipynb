# Matplotlib 之三：条形图 / 柱状图（bar）

---

条形图用来比较**各个离散类别的数值大小**——比如各科分数、各城市人口、各产品销量。每根"柱子"代表一个类别，高度代表数值。

## 学习目标

1. 用 `plt.bar()` 画竖直柱状图、`plt.barh()` 画横向条形图
2. 分清**柱状图（bar）和直方图（histogram）**（初学者极易混）
3. 掌握 `width/height`、`color`、数值标注等参数
4. 会画"并列柱状图"比较多组数据
5. 再次巩固中文显示配置


---
## 一、什么时候用条形图？柱状图 ≠ 直方图！

这是初学者最容易搞混的一对，必须分清：

| 对比 | 柱状图 `bar` | 直方图 `hist` |
|---|---|---|
| x 轴 | **离散的类别**（语文/数学/英语） | **连续的数值区间**（0~10, 10~20…） |
| 柱子之间 | 通常有间隙 | 通常紧挨着（连续） |
| 回答的问题 | 各类别谁多谁少 | 数据整体"分布长啥样" |
| 数据来源 | 已经汇总好的类别值 | 一堆原始数据，自动分箱统计 |

一句话：**类别对比用 `bar`，看分布形状用 `hist`**（直方图本模块没单独讲，但记住它俩不是一回事）。


---
## 二、最基础的竖直柱状图

`plt.bar(x, height)`：`x` 是类别位置（或类别名），`height` 是每根柱子的高度。

> **原稿问题**：原稿图例写成 `plt.legend(["baby"])`——"baby"和"成绩"毫不相干，纯属误写。图例文本应当和图要表达的内容一致，下面改成 `"分数"`。另外原稿在 `for` 循环后面挂了一大段 `"""..."""` 多行字符串当"注释"，虽然不会报错，但把说明塞进代码里是坏习惯，改成 markdown 讲解更清楚。


````python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

subjects = ['Math', 'Physics', 'Chemistry', 'Biology', 'History']
scores = [80, 90, 75, 85, 95]

plt.figure(figsize=(10, 8), dpi=100)
plt.bar(subjects, scores, color='skyblue', width=0.5)

plt.title('Student Scores')
plt.xlabel('Subjects', fontsize=16)
plt.ylabel('Scores', fontsize=16)
plt.legend(["分数"], loc='upper left')                 # 原稿写 "baby"，已修正
plt.grid(axis='y', linestyle="--", alpha=0.3, color="red")
plt.ylim(0, 100)

# 在每根柱子顶端标数值：y 坐标用柱子高度，va="bottom" 让字在上方
for x, y in zip(subjects, scores):
    plt.text(x, y, y, fontsize=12, ha="center", va="bottom")

plt.tight_layout()   # 自动调整边距，避免标签被切掉
plt.show()           # 输出: 5 根天蓝色柱子，顶端标着分数
````

## 三、中文显示问题（本模块必讲，再巩固一次）

柱子的类别名、标题一旦带中文，照样变方块。原理不变：**默认字体不含中文**。

```python
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False
```

各系统字体：Windows 用 `SimHei`/`Microsoft YaHei`；macOS 用 `Arial Unicode MS`；Linux 用 `WenQuanYi Zen Hei`。末尾 `DejaVu Sans` 兜底。


````python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

plt.figure(figsize=(8, 5))
plt.bar(['苹果', '香蕉', '橘子'], [30, 45, 25])
plt.title('水果销量')
plt.xlabel('水果'); plt.ylabel('销量（斤）')
plt.show()   # 输出: 中文类别名和标题都正常
````

## 四、横向条形图 `barh` 与核心参数

类别名很长、或类别很多时，竖着放会挤，用 `plt.barh(y, width)` 横着画更清楚。**注意参数名变了**：竖直是 `width`（柱宽），横向是 `height`（柱高，这里指粗细）。

| 参数 | 竖直 `bar` | 横向 `barh` |
|---|---|---|
| 主数值参数 | `height`（柱高） | `width`（柱长） |
| 粗细参数 | `width` | `height` |
| x 轴含义 | 类别 | 数值 |
| y 轴含义 | 数值 | 类别 |


````python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

subjects = ['Math', 'Physics', 'Chemistry', 'Biology', 'History']
scores = [80, 90, 75, 85, 95]

plt.figure(figsize=(10, 8), dpi=100)
plt.barh(subjects, scores, color='skyblue', height=0.5)   # height 相当于竖直图的 width
plt.title('Student Scores')
plt.xlabel('Scores', fontsize=16)
plt.ylabel('Subjects', fontsize=16)
plt.legend(["分数"], loc='upper left')
plt.xlim(0, 100)
plt.tight_layout()
plt.show()   # 输出: 横着的 5 根柱子，越长分数越高
````

## 五、并列柱状图（多组数据对比）

要在一张图里对比"两次考试"的成绩，就得画两组柱子并排。关键是给第二组柱子**加一个偏移量**（柱子宽度 `w`），否则两组会重叠。

公式：`x` 位置 = 类别下标；第一组在 `x - w/2`，第二组在 `x + w/2`。


````python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

subjects = ['Math', 'Physics', 'Chemistry', 'Biology', 'History']
mid = [80, 90, 75, 85, 95]
final = [85, 88, 82, 80, 92]

x = np.arange(len(subjects))   # [0,1,2,3,4]
w = 0.35                        # 柱子宽度

plt.figure(figsize=(10, 8), dpi=100)
plt.bar(x - w/2, mid, w, label='期中', color='skyblue')
plt.bar(x + w/2, final, w, label='期末', color='salmon')

plt.xticks(x, subjects)        # 用类别名替代数字下标
plt.ylabel('Scores')
plt.title('期中 vs 期末成绩')
plt.legend()
plt.tight_layout()
plt.show()   # 输出: 每科两根并排柱子，蓝=期中、红=期末
````

## 六、常见坑清单

| 坑 | 现象 | 正解 |
|---|---|---|
| 柱状图当直方图用 | 把原始数据直接喂给 `bar` | 看分布用 `plt.hist` |
| 图例文本瞎写 | 图例和内容对不上 | 图例写真实含义，如"分数" |
| 并列柱体重叠 | 两组柱子盖一起 | 第二组加 `w/2` 偏移 |
| 类别名被切掉 | 边缘标签显示不全 | 加 `plt.tight_layout()` |
| 中文方块 | 标题/类别名变 □ | 设 `font.sans-serif` 中文字体 |
