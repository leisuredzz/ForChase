## 一、matplotlib 是什么？

一句话：**Python 里最主流的"画图"库**。

它的地位有点像绘图界的"老大哥"——`pandas` 自带的图表、`seaborn` 这些更高级的库，
底层都是它。所以**学会 matplotlib，等于拿到了画图的基本功**。

| 你想画什么 | 用哪个函数 | 本模块哪一篇 |
|---|---|---|
| 折线图（看趋势） | `plt.plot()` | `08.02` |
| 散点图（看分布/相关性） | `plt.scatter()` | `08.03` |
| 条形图（比大小） | `plt.bar()` | `08.04` |
| 饼图（看占比） | `plt.pie()` | `08.05` |
| 箱型图（看离群值） | `plt.boxplot()` | `08.06` |
| 多张子图 | `plt.subplots()` | `08.07` |


## 二、第一张图：三步走

画图这件事，拆开来永远是三步：

1. **准备数据**（两组等长的数字：横坐标 `x`、纵坐标 `y`）
2. **画**（`plt.plot(x, y)`）
3. **显示**（`plt.show()`）

就这样，跑一下看看：


````python
import matplotlib.pyplot as plt

# 第 1 步：准备数据
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# 第 2 步：画线
plt.plot(x, y)

# 第 3 步：显示出来
plt.show()   # 输出: 一条从 (1,2) 到 (5,10) 的直线
````

看到那条直线了吗？恭喜，你已经会画图了。

但如果只到这里，你很快就会撞上第一个坑——**中文变成方块**。
在讲那个坑之前，先把两个基础概念理清楚。


## 三、`Figure` 和 `Axes` 是什么？

这两个词在看文档时会反复出现，先建立直觉：

| 名词 | 中文 | 类比 | 怎么拿到 |
|---|---|---|---|
| **Figure** | 画布 | 一整张纸 | `plt.figure()` |
| **Axes** | 坐标系 | 纸上的一块"绘图区域"，含坐标轴、刻度、标题 | `fig, ax = plt.subplots()` |

关键点：

- **一张 Figure 上可以放多块 Axes**（就是"一张纸上画好几张小图"）
- 用 `plt.xxx()` 画图时，matplotlib 内部帮你记着"当前是哪张画布、哪块坐标系"
  （这叫 **pyplot 写法**，适合快速画一张图）
- 要精细控制（多子图）时，就显式拿到 `fig, ax` 对象（这叫**面向对象写法**）

两种写法的详细对比在 `08.02-折线图基础` 那一篇，这里先有个印象即可。


### 顺便认识 `figsize` 和 `dpi`

```python
plt.figure(figsize=(8, 6), dpi=100)
```

| 参数 | 含义 | 注意 |
|---|---|---|
| `figsize=(宽, 高)` | 画布尺寸 | 单位是**英寸**，不是像素！ |
| `dpi` | 每英寸的像素数 | 越大越清晰、文件越大，常用 80~120 |

所以 `figsize=(8, 6), dpi=100` 实际画出来大约是 `800 x 600` 像素。


````python
plt.figure(figsize=(8, 6), dpi=100)   # 8英寸 x 6英寸，每英寸100像素
plt.plot([1, 2, 3, 4, 5], [2, 4, 6, 8, 10])
plt.show()   # 输出: 与上一张形状相同，但更宽更高
````

## 四、头号坑：`plt.show` 少了括号

这是初学者**几乎必然踩一次**的坑，值得单独看一遍：

```python
plt.show      # 错！这只是"拿到了函数本身"，什么都不会发生
plt.show()    # 对！带括号才是"调用它"
```

为什么？因为在 Python 里，**函数名不加括号就是一个"函数对象"**，加括号才是执行。
你可以把函数想象成一张菜谱：`plt.show` 是"看着菜谱"，`plt.show()` 才是"照着做菜"。

更坑的是：**它不报错**。代码安安静静跑完，图就是不出现——所以特别难查。

下面把"函数对象长什么样"打出来给你看，这样你以后一眼就能认出来：


````python
print('不加括号拿到的是:', plt.show)
print('类型是:', type(plt.show))
print()
print('加括号才是调用它（下面这行会真的把图显示出来）：')
plt.figure()
plt.plot([1, 2, 3], [3, 2, 1])
plt.show()   # 输出: 图正常显示
````

> **同类陷阱**：任何函数少写括号都可能出这种问题，比如 `plt.legend`（少括号图例不显示）、
> `df.info`（少括号什么都没输出）。养成习惯：**看到"没反应又不报错"，先检查括号**。


## 五、头号坑的第二集：中文显示成方块

现在试着给图加个中文标题：


````python
plt.figure()
plt.plot([1, 2, 3], [1, 4, 9])
plt.title('中文标题测试')     # 输出: 大概率是一堆方块 □□□□□□
plt.show()
````

看到方块了？原因很简单：

**matplotlib 的默认字体 `DejaVu Sans` 里根本没有中文字形。**
遇到中文它找不到字模，就画成方块（这叫"豆腐块"）。

解决办法只有一条：**告诉 matplotlib 换一款带中文的字体**。
而且还得顺带处理负号——因为很多中文字体里连 `-` 也缺字模：

| 配置项 | 作用 |
|---|---|
| `plt.rcParams['font.sans-serif']` | 无衬线字体列表，**按顺序找第一个装了的** |
| `plt.rcParams['axes.unicode_minus'] = False` | 让负号走 ASCII，避免也变方块 |

不同系统自带的中文字体不一样，**给一个列表兜底**最稳：

| 系统 | 可用中文字体 |
|---|---|
| Windows | `SimHei`（黑体）、`Microsoft YaHei`（雅黑）、`KaiTi`、`FangSong` |
| macOS | `Arial Unicode MS`、`PingFang SC`、`Heiti SC` |
| Linux | `WenQuanYi Zen Hei`、`Noto Sans CJK SC` |


这段配置**建议每个需要中文的 notebook 开头都贴一次**，全项目通用：


````python
import matplotlib.pyplot as plt

# 【关键】一次性配置：支持中文 + 负号
plt.rcParams['font.sans-serif'] = [
    'SimHei',            # 黑体（Windows 常见）
    'Microsoft YaHei',   # 微软雅黑
    'KaiTi',             # 楷体
    'FangSong',          # 仿宋
    'Arial Unicode MS',  # macOS 常见
    'DejaVu Sans',       # 兜底：没有中文，但保证不崩
]
plt.rcParams['axes.unicode_minus'] = False   # 负号正常显示

plt.figure()
plt.plot([1, 2, 3], [1, 4, 9])
plt.title('中文标题现在正常了')      # 输出: 中文正常显示，不再是方块
plt.xlabel('横坐标')
plt.ylabel('纵坐标')
plt.show()
````

## 六、把图保存成图片

用 `plt.savefig('文件名.png')`。这里有一个**必须记住的顺序**：

```python
plt.savefig('图.png')   # 必须在前面
plt.show()              # show 会清空当前画布，放后面才安全
```

**原因**：`plt.show()` 之后当前的画布状态会被重置，你再调用 `savefig`
有可能存出一张全白的图。所以规矩就是：**先存后显**。


````python
import os

OUT_DIR = os.path.join(os.getcwd(), '生成文件')
os.makedirs(OUT_DIR, exist_ok=True)          # 产物统一放 生成文件/，不往项目根目录乱丢

plt.figure(figsize=(6, 4), dpi=100)
plt.plot([1, 2, 3, 4, 5], [2, 4, 6, 8, 10], marker='o')
plt.title('我的第一张图')
plt.xlabel('x')
plt.ylabel('y')

png = os.path.join(OUT_DIR, 'helloworld.png')
plt.savefig(png)          # 【顺序】先 savefig
plt.show()                # 【顺序】后 show

print('已保存到：', png)
print('文件大小：', os.path.getsize(png), '字节')
````

## 七、小结与常见坑清单

画图三步：**准备数据 -> `plt.plot()` -> `plt.show()`**。

| 坑 | 现象 | 正解 |
|---|---|---|
| `plt.show` 少括号 | 不报错，但图不出现 | 写成 `plt.show()` |
| 中文显示成方块 □ | 标题/标签全是方块 | 配 `font.sans-serif` 中文字体列表 |
| 负号显示成方块 | `-5` 变方块 | 加 `axes.unicode_minus = False` |
| `savefig` 存出白图 | 图片全白 | `savefig` 写在 `show` **之前** |
| 忘了 `plt.figure()` | 线画到上一张图上 | 每张新图先 `plt.figure()` |
| `figsize` 以为是像素 | 图尺寸和预期差很多 | `figsize` 单位是**英寸**，配合 `dpi` |
