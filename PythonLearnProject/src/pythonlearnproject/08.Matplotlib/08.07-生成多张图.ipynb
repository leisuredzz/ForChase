# Matplotlib 之六：一张图里画多张子图

---

前面每篇都只画一张图。实际汇报常常要把**折线、柱状、散点、饼图**摆在一页上对比。这一篇讲两种排布多张子图的方法，并和前面学的"pyplot vs 面向对象"串起来。

## 学习目标

1. 用 `plt.subplot(行, 列, 第几个)` 逐个往网格里塞图
2. 用 `plt.subplots(行, 列)` 一次性拿到所有坐标轴，面向对象地画
3. 搞清这两种写法的区别与取舍
4. 把前面学过的四种图合进同一页
5. 再次巩固中文显示配置


---
## 一、两种思路对比

| 写法 | 关键调用 | 特点 |
|---|---|---|
| 逐个建 | `plt.subplot(2,2,1)` 返回第 1 个 axes | 一次只拿到"当前"那张子图，循环/顺序调用 |
| 一次建 | `fig, axes = plt.subplots(2,2)` 返回整网格 | 一次性拿到所有 axes（数组），用 `axes[i,j]` 访问，更整齐 |

简单说：`subplot` 是"建一张用一张"，`subplots` 是"先把 2×2 的画布格子全建好，再挨个填"。**子图多时 `subplots` 更清爽**。


---
## 二、方法 A：`plt.subplot` 逐个建

`plt.subplot(nrows, ncols, index)`：把画布分成 `nrows×ncols` 格，`index` 从 1 开始、按行数（左上=1）。它返回当前这个格子的 axes 对象，后面直接用 `f1.plot(...)` 之类往里画。


````python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

month = [f"{i}月" for i in range(1, 8)]
sale = [100, 120, 150, 180, 200, 180, 150]
temp = [17, 17, 18, 19, 20, 18, 15]

plt.figure(figsize=(12, 8), dpi=100)

# 第 1 格：折线图
f1 = plt.subplot(2, 2, 1)
f1.plot(month, sale, label="销售额", marker="o", linestyle="--",
        linewidth=2, color="purple")
f1.set_title("销售额折线图")
f1.set_xlabel("月份"); f1.set_ylabel("销售额")
f1.legend()

# 第 2 格：柱状图
f2 = plt.subplot(2, 2, 2)
f2.bar(month, sale, label="销售额", color="lightblue")
f2.set_title("销售额柱状图")
f2.set_xlabel("月份"); f2.set_ylabel("销售额")
f2.legend()

# 第 3 格：散点图
f3 = plt.subplot(2, 2, 3)
f3.scatter(month, temp, label="温度", marker="o", color="purple", s=100)
f3.set_title("温度散点图")
f3.set_xlabel("月份"); f3.set_ylabel("温度")
f3.legend()

# 第 4 格：饼图
f4 = plt.subplot(2, 2, 4)
f4.pie(sale, labels=month, autopct="%1.1f%%", startangle=90,
       colors=["#FF9999", "#66B2FF", "#99FF99", "#FFCC99",
               "#FF99FF", "#CC99FF", "#FF6666"])
f4.set_title("销售额占比饼图")

plt.tight_layout()   # 自动调整子图间距，防止标签重叠
plt.show()           # 输出: 2×2 共 4 张子图排在一页
````

---
## 三、方法 B：`plt.subplots` 一次性建（更推荐）

`fig, axes = plt.subplots(2, 2)` 返回一个 `figure` 和一个 `2×2` 的 axes 数组。`axes[0,0]` 就是左上角那格。所有子图共用一个 `fig`，排版由 `plt.tight_layout()` 统一处理。这种写法一眼能看全布局，最适合"图表多"的场景。


````python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

month = [f"{i}月" for i in range(1, 8)]
sale = [100, 120, 150, 180, 200, 180, 150]
temp = [17, 17, 18, 19, 20, 18, 15]

# 一次性拿到画布 + 2×2 的 axes 网格
fig, axes = plt.subplots(2, 2, figsize=(12, 8), dpi=100)

axes[0, 0].plot(month, sale, label="销售额", marker="o",
                linestyle="--", linewidth=2, color="purple")
axes[0, 0].set_title("销售额折线图")
axes[0, 0].set_xlabel("月份"); axes[0, 0].set_ylabel("销售额")
axes[0, 0].legend()

axes[0, 1].bar(month, sale, label="销售额", color="lightblue")
axes[0, 1].set_title("销售额柱状图")
axes[0, 1].set_xlabel("月份"); axes[0, 1].set_ylabel("销售额")
axes[0, 1].legend()

axes[1, 0].scatter(month, temp, label="温度", marker="o",
                   color="purple", s=100)
axes[1, 0].set_title("温度散点图")
axes[1, 0].set_xlabel("月份"); axes[1, 0].set_ylabel("温度")
axes[1, 0].legend()

axes[1, 1].pie(sale, labels=month, autopct="%1.1f%%", startangle=90,
               colors=["#FF9999", "#66B2FF", "#99FF99", "#FFCC99",
                       "#FF99FF", "#CC99FF", "#FF6666"])
axes[1, 1].set_title("销售额占比饼图")

plt.tight_layout()
plt.show()   # 输出: 与方法 A 同样的 2×2 布局，但代码更整齐
````

## 四、中文显示问题（本模块必讲，最后巩固一次）

多子图里每一格的标题、轴标签、饼图 labels 都可能带中文，同样会变方块。配置和前几篇一模一样：

```python
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False
```

Windows 用 `SimHei`/`Microsoft YaHei`；macOS 用 `Arial Unicode MS`；Linux 用 `WenQuanYi Zen Hei`。`axes.unicode_minus=False` 不要漏。

## 五、常见坑清单

| 坑 | 现象 | 正解 |
|---|---|---|
| `subplot` 的 index 搞错 | 图放错格 | index 从 1 开始、按行数（左上=1） |
| 子图标签重叠 | 挤在一起 | 末尾加 `plt.tight_layout()` |
| `subplots` 当 `subplot` 用 | 拿到的不是单 axes | 前者返回 `(fig, axes数组)` |
| 多图画串 | 新图叠到旧图上 | 每张子图前 `plt.subplot(...)` / 用独立 axes |
| 中文方块 | 子图标题变 □ | 设 `font.sans-serif` 中文字体 |

## 六、动手练习

**练习 1**：用 `plt.subplots(1, 2)` 画左折线、右柱状两张子图，数据自定。

**练习 2**：用 `plt.subplots(2, 1)` 上下排两张图，上面 `sin`、下面 `cos`。

**练习 3**：把"箱型图"也加进来，做一个 `2×2` 布局，包含折线/柱状/散点/箱型。

**练习 4**：给多子图整体加一个总标题 `fig.suptitle("我的可视化大屏")`。

---

**答案思路（先自己做）**

```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# 1
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot([1,2,3], [1,4,2]); axes[0].set_title("折线")
axes[1].bar([1,2,3], [3,1,2]); axes[1].set_title("柱状")
plt.tight_layout(); plt.show()

# 2
t = np.linspace(0, 2*np.pi, 100)
fig, axes = plt.subplots(2, 1, figsize=(6, 6))
axes[0].plot(t, np.sin(t)); axes[1].plot(t, np.cos(t))
plt.tight_layout(); plt.show()

# 4 总标题
fig, axes = plt.subplots(1, 2)
axes[0].plot([1,2,3],[1,2,3]); axes[1].plot([1,2,3],[3,2,1])
fig.suptitle("我的可视化大屏"); plt.tight_layout(); plt.show()
```
