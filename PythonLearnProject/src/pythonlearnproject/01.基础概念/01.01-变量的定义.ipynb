# 第 2 课 · 变量的定义 —— 给数据起个名字

### 学习目标
1. 理解"变量"到底是什么（不是盒子，是标签）
2. 分清 = 和 == 这两个符号
3. 知道一个变量身上同时带着 4 样东西

### 一句话理解
变量不是"装东西的盒子"，而是"贴在数据上的便利贴"。
你写 a = 10，意思是：把便利贴 a 贴到 10 这个数据上。


## 1. 赋值号 = ：把右边的东西，让左边这个名字指向它


```python
a = 10              # 让名字 a 指向数字 10
print(a)            # 输出 10

# 【千万注意】= 不是数学里的"等于"！
#   a = 10   是"赋值"：把 10 交给 a
#   a == 10  是"判断"：问一句"a 是不是等于 10"，结果是 True / False
print(a == 10)      # 输出 True  —— 这是比较运算，不是赋值
print(a == 8)       # 输出 False
```

## 2. 一个变量身上，同时带着 4 样东西

以 a = 10 为例：
  1) 变量的名称      a            —— 你用来称呼它的名字
  2) 变量保存的数据  10            —— 真正被存下来的值
  3) 变量的数据类型  int          —— 决定它能参与哪些运算
  4) 变量的内存地址  0x...        —— 数据在内存里的门牌号

前三个用 type() 和 id() 就能看到：


```python
print(type(a))      # <class 'int'>
print(id(a))        # 一串数字，就是它的内存地址（每次运行可能不同）
```

## 3. 变量是"标签"，不是"盒子"——这是理解 Python 的关键一步

很多人以为 a = 10 是"拿出一个叫 a 的盒子，把 10 放进去"。
Python 的真实做法是：先在内存里创建 10，再把标签 a 贴到它身上。

这解释了两个初学者容易懵的现象：

现象 1：重新赋值 = 换一张标签，原来的数据还在（只是没人用它了）


```python
a = 20              # 现在是 a 这张标签改贴到 20 上了
print(a)            # 输出 20

# 现象 2：两个名字可以贴到同一个数据上
b = a               # b 和 a 现在指向同一个数据
print(a, b)         # 20 20
b = 30              # b 换了目标，a 不受影响（因为这里换的是"标签"，不是"数据"）
print(a, b)         # 20 30

# 【对照】列表这种"可变对象"就不一样了：改内容时两边都会变
list_a = [1, 2, 3]
list_b = list_a     # 两个标签贴到【同一个列表】上
list_b.append(4)
print(list_a)       # [1, 2, 3, 4]  —— a 也变了！
# 原因：append 改的是"数据本身"，两个标签指向的还是同一份数据。
# 想让它们互不影响，要显式复制：list_c = list_a.copy()
```

## 4. 变量的生命周期

变量在【第一次被赋值】时诞生（此时才在内存中划分空间并初始化）。
之后再次赋值，不会重新划分空间，只是修改指向的内容。


```python
print(hasattr(__builtins__, 'sorted'))   # True，证明内置函数一直存在

# 查看当前环境里已经定义了哪些名字（初学者调试时挺有用）
# print(dir())      # 取消注释可以看到一长串名字
```

## 常见坑

坑 1：用了没定义过的变量
    print(score)   —— NameError: name 'score' is not defined
    解决：先赋值，再使用。

坑 2：把 = 和 == 写反
    if a = 10:     —— SyntaxError，if 里要的是判断，不是赋值
    if a == 10:    —— 正解

坑 3：以为 b = a 之后改 b 会连带改 a
    对数字、字符串成立（不可变类型），对列表、字典不成立（可变类型）。
    记住一句话：可变对象"传的是同一个东西"，不可变对象"传的是值"。
