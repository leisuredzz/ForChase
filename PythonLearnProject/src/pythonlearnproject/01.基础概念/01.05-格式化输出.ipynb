# 第 6 课 · 格式化输出 —— 把变量"嵌"进文字里

### 学习目标
1. 掌握 3 代格式化写法：% 占位符、str.format()、f-string
2. 知道每代的坑在哪，为什么新代码一律用 f-string
3. 掌握 f-string 的常用格式控制（小数位、百分比、千分位、对齐）

### 一句话理解
格式化输出就是"填空题"：先写好带空格的句子，再把变量的值填进去。


## 第 1 代：% 占位符（最老，沿袭自 C 语言）

语法：'模板 % 格式符' % 参数

常用格式符速查：
  %s   字符串（几乎所有东西都能塞进去）
  %d   有符号十进制整数
  %f   浮点数（默认保留 6 位小数）
  %c   单个字符
  %o   八进制整数
  %x   十六进制整数
  %e   科学计数法（小写 e）
  %E   科学计数法（大写 E）
  %g   %f 和 %e 的自动选择（按数值大小挑更短的那个）
  %%   输出一个真正的百分号（因为 % 本身是特殊符号，要写两个）


```python
name = "Wilia"
age = 18
result = '我的姓名是%s，年龄是%d' % (name, age)
print(result)

# 多个参数要用【元组】包起来，顺序必须和占位符一一对应
print("My name is %s and I am %d years old." % ("Alice", 30))

# 控制小数位数：%.2f 表示保留 2 位小数（会自动四舍五入）
a = 1.23456
print('%f' % a)          # 1.234560  —— 不指定就给你 6 位
print('%.2f' % a)        # 1.23
print('%.4f' % a)        # 1.2346    —— 注意最后一位进位了

# 八进制、十六进制
d = 9
print('八进制：%o' % d)      # 11
print('十六进制：%x' % d)    # 9

# 【重点坑】%d 遇到浮点数会直接截断（不是四舍五入！）
c = 10.45
print("被 %d 处理后的结果：%s" % (c, "%d" % c))     # 10，小数点后全丢掉
print("对 10.99 用 %d：", "%d" % 10.99)             # 10，不是 11！
# 所以：想保住小数就用 %f 或 %.2f，别用 %d

# 第 1 代的缺点：参数多了以后，顺序特别容易写错，而且完全看不出哪个占位符配哪个变量
print('I want {1} pieces of item {0} for {2} dollars.'.format(3, 567, 49.95))
```

## 第 2 代：str.format() —— 用 {} 占位，可以点名


```python
print("My name is {} and I am {} years old.".format("Bob", 25))

# 带下标的写法：{0} 表示第 0 个参数，{1} 表示第 1 个（可以重复用、可以换顺序）
print('{0}，{1}，再见了{0}'.format('张三', '李四'))     # 张三，李四，再见了张三

# 带名字的写法：可读性最好，参数顺序无所谓了
print('{name} 今年 {age} 岁，来自 {city}'.format(name='Cedric', age=22, city='武汉'))

# 直接复用格式控制（和 % 那套类似）
print('{:.2f}'.format(3.14159))          # 3.14
print('{:>10}'.format('abc'))            # 右对齐，总宽 10
```

## 第 3 代：f-string —— 当前唯一推荐的写法（Python 3.6+）


```python
name = "Charlie"
age = 28
print(f"My name is {name} and I am {age} years old.")

# 为什么推荐它？
#   1. 变量直接写在原地，不用回头看参数列表，一眼就知道谁是谁
#   2. 大括号里可以写任意表达式，会先算出来再填进去
print(f'明年我 {age + 1} 岁')
print(f'名字大写：{name.upper()}')
print(f'名字长度：{len(name)}')

# f-string 的格式控制：冒号后面写"格式规格"
pi = 3.14159265
print(f'{pi:.2f}')          # 3.14          保留 2 位小数
print(f'{pi:.0f}')          # 3             不带小数
print(f'{1234567:,}')       # 1,234,567     千分位分隔
print(f'{0.856:.1%}')       # 85.6%         百分比（会自动乘以 100）
print(f'{42:>8}')           #       42      右对齐，宽度 8
print(f'{42:<8}|')          # 42      |     左对齐
print(f'{42:^8}|')          #    42   |     居中
print(f'{42:08d}')          # 00000042      补零

# 格式化输出最经典的用途：对齐表格
students = [('张三', 95, 0.95), ('李四', 88, 0.88), ('王小明', 76, 0.76)]
print(f'{"姓名":<8}{"分数":>6}{"占比":>10}')
print('-' * 24)
for n, s, r in students:
    print(f'{n:<8}{s:>6}{r:>10.1%}')
# 输出会整整齐齐——不需要任何手工空格，宽度由格式规格保证。
# 这里还能看出对齐的细节：中文字符在终端里占 2 个字符宽，但 Python 按 1 个字符算长度，
# 所以出现了中英文混排时会有点错位，需要时可以用 wcwidth 之类的库校准。
```

## 三代写法对比：同一个需求，写三遍


```python
price, qty = 19.9, 3
print('总价：%.2f 元' % (price * qty))                       # 第 1 代
print('总价：{:.2f} 元'.format(price * qty))                  # 第 2 代
print(f'总价：{price * qty:.2f} 元')                          # 第 3 代（推荐）
# 结论：新代码一律用 f-string。只有在维护老项目、或者模板需要动态拼接时才用前两代。
```

## 常见坑

坑 1：% 占位符数量对不上
    '我 %s 岁' % ('x', 'y')     -> TypeError: not all arguments converted
    '我 %s 岁 %s' % ('x',)      -> TypeError: not enough arguments
坑 2：只想输出一个百分号，却写成了 %
    print('折扣 30%')           -> 会报错；正解 '折扣 30%%'
坑 3：f-string 里用了外层的同种引号
    f"他叫{"张三"}"             -> 报错（Python 3.12 之前）；正解用不同的引号
坑 4：%d 当四舍五入用 —— 它是截断，10.99 会变成 10
坑 5：变量名写错，f-string 里就成了普通文字
    f'是 {nane} 啊'             -> NameError，Python 不会帮你猜
