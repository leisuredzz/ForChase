# 第 4 课 · 数值类型 —— Python 里的"数字家族"

### 学习目标
1. 认清 4 种数值类型：int / float / bool / complex
2. 理解为什么 0.1 + 0.2 不等于 0.3
3. 会用 type() 和 isinstance() 判断类型

### 一句话理解
Python 里的数字分四种：整数、小数、真假、复数。
日常写代码 99% 的情况只用前三种。


## 1. int 整数：不带小数点的数


```python
a = 100
b = -7
c = 1_000_000          # 下划线只是给人看的，Python 会自动忽略（大数字可读性技巧）
print('a =', a, type(a))
print('c =', c, '（1_000_000 和 1000000 完全等价）')

# Python 的 int 没有大小上限，想多大就多大（这点和 C/Java 不同）
big = 2 ** 100
print('2 的 100 次方 =', big)
```

## 2. float 浮点数：带小数点的数


```python
f1 = 1.3
f2 = 3.14
print('f1 =', f1, type(f1))

# 科学计数法：3.1E5 读作"3.1 乘以 10 的 5 次方"
#   注意：E 后面的 5 是【指数】，不是位数
e = 3.1E5               # = 3.1 * 10^5 = 310000.0
print('3.1E5 =', e, type(e))

# 【重要】浮点数运算的精度问题 —— 面试和实际开发都会遇到
print('0.1 + 0.2 =', 0.1 + 0.2)          # 0.30000000000000004，不是 0.3！
print('0.1 + 0.2 == 0.3 ?', 0.1 + 0.2 == 0.3)   # False

# 原因：计算机用二进制存小数，而 0.1、0.2 在二进制里是【无限循环小数】，
#      存不下只能截断，截断就会产生误差。这不是 Python 的错，是所有语言的通病。
#
# 解决办法 1：用 round() 四舍五入
print('round 一下：', round(0.1 + 0.2, 2))          # 0.3
# 解决办法 2：涉及金额时用 Decimal（精确十进制运算）
from decimal import Decimal
print('Decimal：', Decimal('0.1') + Decimal('0.2'))  # 0.3
# 解决办法 3：判断相等时不要用 ==，而是看差值是否足够小
print('用差值判断：', abs((0.1 + 0.2) - 0.3) < 1e-9)  # True
```

## 3. bool 布尔型：只有 True 和 False 两个值


```python
print(True)
print(type(True))

# 关键特性：布尔值本质上是整数的子类，True 就是 1，False 就是 0
print('True + 2 =', True + 2)        # 3
print('False + 2 =', False + 2)      # 2
print('True * 10 =', True * 10)      # 10

# 这个特性有很多妙用，比如统计"满足条件的个数"：
scores = [88, 45, 92, 61, 77]
passed = sum(score >= 60 for score in scores)   # 每个条件产生 True/False，加总就是个数
print('及格人数：', passed)          # 4
print('总共 5 人，通过率：', f'{passed / len(scores):.0%}')
```

## 4. complex 复数：a + bj（工程计算和信号处理才用得到）


```python
z = 3 + 4j
print('z =', z, type(z))
print('实部：', z.real, '虚部：', z.imag)
print('模长：', abs(z))              # 5.0，勾股定理 sqrt(3² + 4²)
```

## 5. 怎么判断一个值是哪种类型

方法 1：type(x) —— 直接问"你是什么类型"


```python
print(type(100))        # <class 'int'>
print(type(1.5))        # <class 'float'>

# 方法 2：isinstance(x, 类型) —— 问"你是不是属于这个类型（或它的子类）"
# 推荐用这个，因为它对子类友好（bool 是 int 的子类，所以下面第一个是 True）
print(isinstance(True, int))          # True
print(isinstance(100, int))           # True
print(isinstance(100, float))         # False

# 实战：写一个函数，能同时接受整数和浮点数
def double(x):
    if isinstance(x, (int, float)):
        return x * 2
    return '只支持数字哦'

print(double(10), double(2.5), double('abc'))
```

## 6. 小实战：算 BMI，顺手看看浮点精度


```python
height = 1.75          # 米
weight = 75.0          # 千克
bmi = weight / height ** 2
print(f'BMI = {bmi:.2f}')            # :.2f 表示保留 2 位小数

# 对照一下整数除法的两种结果（后面运算符那课会细讲）
print('7 / 2 =', 7 / 2)              # 3.5  普通除法，结果一定是 float
print('7 // 2 =', 7 // 2)            # 3    整除，向下取整
print('7 % 2 =', 7 % 2)              # 1    取余
```

## 常见坑

坑 1：拿 == 直接比较浮点数
    正解：abs(a - b) < 1e-9，或者干脆用 Decimal
坑 2：以为 / 会得到整数
    7 / 2 永远是 3.5；想要整数用 //
坑 3：大数字里想加分隔符却写了中文逗号
    1，000，000  ->  报错；正解 1_000_000 或 1000000
坑 4：以为 int 会溢出
    Python 里不会；但转成 numpy 的 int32 之后就会（Numpy 那章有专门演示）
