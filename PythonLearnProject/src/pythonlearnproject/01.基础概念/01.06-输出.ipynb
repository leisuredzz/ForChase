# 第 1 课 · 输出 —— 让程序"开口说话"

### 学习目标
1. 会用 print() 输出一个值、多个值
2. 理解 end 和 sep 两个参数到底在控制什么
3. 知道 print 的本质是"把东西变成文字，送到屏幕上"

### 一句话理解
print 就像一个人在念稿子：你给它什么，它就照着念什么。
它不会做计算、不会判断对错，只负责"读出来"。


## 1. 最基础的用法：输出一个值

默认情况下，print 输出完内容会自动换行。
等价于 print('hello world', end='\n')  —— 结尾被补了一个换行符


```python
print('hello world')

# 你能输出任何东西：数字、字符串、列表、甚至表达式的结果
print(100)
print(3 + 5)          # 先算 3+5 得到 8，再把 8 转成文字输出
```

## 2. 输出多个值：用逗号分隔

逗号分隔的多个值，print 会挨个输出，并在中间补一个空格。
注意：这里不是"字符串拼接"，print 会对每个值分别调用 str()


```python
print('孤帆远影碧空尽', '唯见长江天际流')

# 想改掉中间那个空格？用 sep 参数（separator，分隔符）
print('2026', '09', '12', sep='-')     # 输出 2026-09-12
print('C:', 'Users', 'leisuredzz', sep='/')  # 输出 C:/Users/leisuredzz
```

## 3. 控制结尾：end 参数

默认 end='\n'（换行）。把它改成空字符串，下一次 print 就会接在同一行。


```python
print('hello', end='')
print(' world')          # 这两行合起来输出 "hello world"

# 常用场景：循环里不想每次都换行，比如打印乘法表
print('不换行演示：', end='  ')
for i in range(1, 4):
    print(i, end=' ')    # 每次都不换行，只加个空格
print()                  # 循环结束后，手动补一个换行，把"行"收尾
```

## 常见坑

坑 1：中文逗号 / 中文引号
    print("你好“）   —— 报 SyntaxError
    Python 里所有语法符号都必须是【英文半角】的 , " ' ( )

坑 2：忘了引号
    print(hello)     —— NameError，因为 Python 会把 hello 当成一个变量名去找
    print('hello')   —— 正解

坑 3：括号写成了中文括号
    print（"hi"）    —— SyntaxError

坑 4：sep 和 end 是 print 的参数，要写在括号内，不能写在括号外


## 4. 小实战：让 print 拼出一个"小程序界面"


```python
print('=' * 40)          # 字符串乘以整数 = 重复拼接，非常常用的分隔线技巧
print('欢迎使用温度记录小工具'.center(40, '-'))
print('=' * 40)

name = 'Cedric'
city = '武汉'
# f-string：在字符串前面加 f，然后用 {} 把变量"塞"进去（后面专门有一课讲）
print(f'{name} 你好，今天 {city} 的温度是 26 度')
print('=' * 40)
```
