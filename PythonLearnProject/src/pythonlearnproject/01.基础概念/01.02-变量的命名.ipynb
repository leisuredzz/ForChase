# 第 3 课 · 变量的命名 —— 名字起得好，代码读得懂

### 学习目标
1. 记住变量名的 4 条硬性规则（违反就直接报错）
2. 认识 Python 的关键字（这些名字你不能用）
3. 掌握 3 种主流命名风格，知道什么时候用哪种

### 一句话理解
变量名是写给人看的，不是写给电脑看的。
电脑只关心"名字有没有重复"，人关心"这名字到底啥意思"。


## 1. 硬性规则：违反这几条，Python 直接报错

  规则 1：只能由【数字、字母、下划线 _】组成
  规则 2：不能以【数字】开头
  规则 3：不能是 Python 的【关键字】（见下面第 2 节）
  规则 4：【严格区分大小写】，Name 和 name 是两个完全不同的变量

合法示例


```python
book_name = 'Python入门'
user_name2 = 'Cedric'
_private = '这是合法的，虽然不推荐这么写'
中文变量名 = 'Python 3 支持中文变量名，但工程上不推荐，容易踩编码坑'

# 非法示例（取消注释会报 SyntaxError）
# 2user = 'x'        # 数字开头
# user-name = 'x'    # 减号不是合法字符，Python 会理解成"减法"
# user name = 'x'    # 空格不是合法字符
# class = 'x'        # class 是关键字
```

## 2. 关键字：Python 自己留用的名字，你不能拿来当变量名

关键字就像"注册商标"，是语言内部已经占用的词。


```python
import keyword

print('总共有', len(keyword.kwlist), '个关键字：')
print(keyword.kwlist)

# 判断某个名字是不是关键字（写代码时拿不准可以查一下）
print('class 是关键字吗？', keyword.iskeyword('class'))   # True
print('school 是关键字吗？', keyword.iskeyword('school'))  # False

# 【小技巧】写变量名时如果拿不准，IDE 会把关键字标成特殊颜色，一眼就能看出来。
```

## 3. 三种命名风格：团队里通常只选一种，保持一致

  风格 1：大驼峰 MyName        —— 每个单词首字母大写（通常用于【类名】）
  风格 2：小驼峰 myName        —— 第一个单词小写，后面首字母大写
  风格 3：下划线 my_name       —— 全小写，单词间用下划线（Python 官方推荐，用于变量和函数）

PEP 8（Python 官方代码风格指南）的结论是：
  变量名、函数名 -> 小写下划线（snake_case）
  类名          -> 大驼峰（PascalCase）
  常量          -> 全大写下划线（MAX_SIZE）


```python
my_name = 'Cedric'          # Python 最推荐的写法
print(my_name)

Schoolname = 'WBU'          # 合法，但不建议：单看名字看不出是几个单词
school_name = 'WBU'         # 更好：单词边界一目了然

MAX_CONNECTIONS = 100       # 常量约定：全大写，提醒别人"这是不要随便改的值"
```

## 4. 【修 bug 现场】缩进混用 —— 初学者最常见的报错之一

原始笔记里下面这个类会直接报错：

    class student():
        name = None
    \tage = None      <- 这一行开头是 Tab
        tel = None      <- 这一行开头是 4 个空格

报错信息：TabError: inconsistent use of tabs and spaces in indentation

原因：Python 用缩进表示"代码的层级关系"，所以对缩进极其敏感。
      Tab 在编辑器里看起来可能等于 4 个空格，但它们是两个不同的字符，
      Python 3 不允许同一个代码块里混着用，直接报错。

正确做法：整个项目统一用【4 个空格】缩进。
         VS Code 里打开设置搜索 "insertSpaces"，勾选后按 Tab 会自动转成 4 空格。
         也可以按下 Ctrl+Shift+P 输入 "Convert Indentation to Spaces" 一键转换。


```python
class Student:
    # 用类属性先占个位，表示"这个类有这几个字段"
    name = None
    age = None
    tel = None

    def __init__(self, name, age, tel):
        # self.name 指的是"这个对象自己的 name"
        self.name = name
        self.age = age
        self.tel = tel
        print('Student 自动创建了一个对象：', self.name)


stu = Student('周杰伦', 31, '1354564654654')
print(stu.name, stu.age, stu.tel)
```

## 常见坑

坑 3：大小写笔误 —— NameError
    userName = 'x'
    print(username)      # 大小写不一致，找不到
坑 4：变量名太随意 —— 不报错，但三个月后你自己看不懂
