# 第 1 题：数组变形与统计

**题目**：创建一个 1 到 12 的数组，把它变形为 3 行 4 列的二维数组，然后：

1. 计算每行的平均值
2. 计算每列的平均值
3. 把数组展平回一维

**知识点**：`np.arange` / `reshape` / `mean(axis=?)` / `flatten`


````python
# 先自己写一版，再往下看
# 提示：np.arange(1, 13) 得到 1~12
````

## 参考解法


````python
arr = np.arange(1, 13)
print('原始一维数组：', arr)

reshaped = arr.reshape(3, 4)
print('\n变形为 3 行 4 列：')
print(reshaped)
````

**第 1 问：每行的平均值** -> 要"消掉列" -> `axis=1`


````python
print('每行平均值（axis=1）：', reshaped.mean(axis=1))
print('  <- 长度 3，对应 3 行')
````

**第 2 问：每列的平均值** -> 要"消掉行" -> `axis=0`


````python
print('每列平均值（axis=0）：', reshaped.mean(axis=0))
print('  <- 长度 4，对应 4 列')
````

**第 3 问：展平** -> `flatten()`（返回副本）或 `ravel()`（返回视图）


````python
flat = reshaped.flatten()
print('展平后：', flat)
print('形状：', flat.shape)
````

> **顺带一提**：`flatten()` 和 `ravel()` 都能展平，区别是
> `flatten()` 返回**副本**（改了不影响原数组），
> `ravel()` 返回**视图**（改了会影响原数组）。
> 只想读、想省内存，用 `ravel()`。


````python
print('ravel 效果一样，但它是视图：', reshaped.ravel())
````

# 第 2 题：随机数组的筛选与替换

**题目**：生成一个 `(5, 5)` 的随机整数数组，取值范围 `[0, 20)`。

1. 找出数组中所有大于 10 的元素

# 第 3 题：销售数据分析

**题目**：某公司六个月的销售额（万元）为 `[120, 135, 110, 125, 130, 140]`：

1. 计算销售额的总和、均值、方差
2. 找出销售额**大于均值**的月份
3. 找出销售额**最高**和**最低**的月份

**知识点**：`sum` / `mean` / `var` / `where` / `argmax` / `argmin`


````python
# 先自己写
````

## 参考解法


````python
moneys = np.array([120, 135, 110, 125, 130, 140])
print('六个月销售额：', moneys)
print('对应月份：     ', np.arange(1, 7))
````

### 第 1 问：总和、均值、方差


````python
total = np.sum(moneys)
mean = np.mean(moneys)
variance = np.var(moneys)

print('总和：', total, '万元')
print('均值：', mean, '万元')
print('方差：', variance)
print('标准差：', np.std(moneys), '  <- 和方差一起看：方差是"平方后的离散度"，标准差开方后回到"万元"单位')
````

### 第 2 问：大于均值的月份

**思路**：先用 `np.where` 拿到**下标**，再 `+1` 变成"月份"（因为下标从 0 开始）。


````python
above = np.where(moneys > mean)[0]        # [0] 是因为 where 返回的是元组
print('大于均值的下标：', above)
print('大于均值的月份：', above + 1, '  <- +1 因为下标从 0 开始，月份从 1 开始')

# 顺便看看是哪些值
print('对应的销售额：  ', moneys[above])
````

### 第 3 问：最高 / 最低的月份

`np.argmax` 返回**最大值所在的下标**（arg = argument = 参数/下标）。


````python
max_idx = np.argmax(moneys)
min_idx = np.argmin(moneys)

print(f'最高销售额：{moneys[max_idx]} 万元，在第 {max_idx + 1} 月')
print(f'最低销售额：{moneys[min_idx]} 万元，在第 {min_idx + 1} 月')
````

> **`max` 和 `argmax` 的区别**：
> `max` 给你**值**，`argmax` 给你**位置**。
> 面试和实际工作中，`argmax` 用得非常频繁（"哪天卖得最好"、"哪个商品最畅销"）。

# 第 4 题：数组的合并与统计

**题目**：给定 `a = [1, 2, 3]`、`b = [4, 5, 6]`：

1. 把 a 和 b **水平**合并
2. 把 a 和 b **垂直**合并
3. 计算合并后数组的均值和标准差

**知识点**：`hstack` / `vstack` / `concatenate` / `mean` / `std`


````python
# 先自己写
````

## 参考解法


````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print('a =', a, ' 形状', a.shape)
print('b =', b, ' 形状', b.shape)
````

### 第 1 问：水平合并

对**一维数组**来说，"水平合并"就是把两串接起来：


````python
horizontal = np.hstack((a, b))
print('hstack 水平合并：', horizontal, ' 形状', horizontal.shape)

# concatenate 等价（一维时效果一样）
c = np.concatenate((a, b))
print('concatenate 等价：', c)
````

> **注意**：`np.hstack((a, b))` 的参数是**元组** `(a, b)`，不是 `a, b`。
> 写成 `np.hstack(a, b)` 会报错。


### 第 2 问：垂直合并

一维数组"垂直叠"就变成了二维（2 行 3 列）:


````python
vertical = np.vstack((a, b))
print('vstack 垂直合并：')
print(vertical)
print('形状：', vertical.shape, '  <- 变成了 2 行 3 列')
````

### 第 3 问：均值和标准差


````python
print('合并后（垂直，2×3）的均值：', np.mean(vertical))
print('合并后（垂直，2×3）的标准差：', np.std(vertical))

# 注意：垂直合并之后元素还是 1~6，所以和水平合并的统计结果一样
print('\n水平合并的均值：', np.mean(horizontal), '（一样，因为元素相同，只是"排法"不同）')
````

> **思考题**：什么情况下"水平和垂直合并的统计结果会不同"？
> 答案是 **一样，永远一样** —— 因为合并只是"排列方式"变了，元素集合没变。
> 但**形状**不同，会让 `axis` 运算的结果完全不同。


````python
print('垂直后 axis=0 求平均（压行）：', vertical.mean(axis=0))
print('垂直后 axis=1 求平均（压列）：', vertical.mean(axis=1))
````

# 第 5 题：去重与计数

**题目**：给定数组 `[2, 1, 2, 3, 1, 4, 3]`：

1. 找出所有唯一的元素
2. 计算每个唯一元素出现的次数

**知识点**：`np.unique` / `return_counts`


````python
# 先自己写
````

## 参考解法


````python
arr = np.array([2, 1, 2, 3, 1, 4, 3])
print('原数组：', arr)
````

`return_counts=True` 是关键的开关：它会**额外返回一个"次数"数组**。


````python
unique_elements, counts = np.unique(arr, return_counts=True)
print('唯一元素（已自动排序）：', unique_elements)
print('对应出现次数：          ', counts)
````

### 变着花样输出结果


````python
# 方法 1：循环逐条打印
print('方法 1 —— 循环：')
for v, c in zip(unique_elements, counts):
    print(f'  元素 {v} 出现了 {c} 次')

# 方法 2：用 np.unique 的返回值构造"频率表"（很像 pandas 的 value_counts）
print('\n方法 2 —— 频率表：')
for i in range(len(unique_elements)):
    bar = '█' * counts[i]                     # 用方块画个简易条形图
    print(f'  {unique_elements[i]}: {counts[i]} 次  {bar}')
````

> **这就是 pandas `value_counts()` 的底层原理** ——
> 后面学 pandas 时你会看到它，但你已经知道它内部是怎么算的了。

# 第 6 题：利润分析

**题目**：某商店 5 天的销售额和成本（万元）如下：

```
销售额：[200, 220, 250, 230, 240]
成本：  [150, 160, 170, 160, 180]
```

1. 计算每天的利润
2. 计算利润的平均值和标准差
3. 找出利润最高的**天数**

**知识点**：数组减法（元素级）/ `mean` / `std` / `argmax`


````python
# 先自己写
````

## 参考解法


````python
sales = np.array([200, 220, 250, 230, 240])
costs = np.array([150, 160, 170, 160, 180])
days = np.arange(1, 6)

print('销售额：', sales)
print('成本：  ', costs)
print('天数：  ', days)
````

### 第 1 问：每天的利润

**一行搞定** —— 这就是 Numpy 元素级运算的威力。
如果用 Python list，你得写循环。


````python
profits = sales - costs
print('每天利润：', profits, '  <- 直接减，逐个元素相减')
````

### 第 2 问：利润的平均值和标准差


````python
print('利润均值：  ', np.mean(profits))
print('利润标准差：', np.std(profits))
print('  <- 标准差小说明利润稳定；大说明波动大')
````

### 第 3 问：利润最高的天数


````python
best = np.argmax(profits)
print(f'利润最高的天数：第 {best + 1} 天，利润 {profits[best]} 万元')

# 顺便看看完整排名（用 argsort）
rank = np.argsort(profits)[::-1]              # 从大到小的下标
print('\n利润从高到低：')
for i in rank:
    print(f'  第 {days[i]} 天：{profits[i]} 万元')
````

# 第 7 题：成绩分档

**题目**：生成 15 个人的随机成绩（0~100），按下面规则分档，并统计各档人数：

| 分数 | 等级 |
|---|---|
| >= 90 | 优秀 |
| >= 75 | 良好 |
| >= 60 | 及格 |
| 其他 | 不及格 |

**知识点**：`np.select` / `np.unique`


````python
# 先自己写
````

## 参考解法


````python
np.random.seed(42)
scores = np.random.randint(0, 101, 15)
print('成绩：', scores)
````

**关键**：`np.select` 的条件要**从严格到宽松**排。


````python
grades = np.select(
    [scores >= 90, scores >= 75, scores >= 60],
    ['优秀', '良好', '及格'],
    default='不及格',
)
print('\n等级：', grades)
````

### 统计各档人数


````python
levels, counts = np.unique(grades, return_counts=True)
print('分档统计：')
for lv, cnt in zip(levels, counts):
    print(f'  {lv}：{cnt} 人')
````

> **为什么 `>= 90` 一定写在最前面？**
> 因为 `select` 是"从上往下第一个匹配就生效"。
> 如果 `>= 60` 写在前面，一个 95 分的人会先被 `>= 60` 抓住判成"及格"，
> "优秀"就永远选不出来了。**顺序就是从紧到松。**

# 第 8 题：数据标准化（Z-Score）

**题目**：有一个 `(4, 3)` 的数据矩阵，**按行**做标准化：

```
z = (x - 该行均值) / 该行标准差
```

做完后验证：每行的均值接近 0，标准差接近 1。

**知识点**：广播 / `mean(axis=1)` / `std(axis=1)` / `reshape(-1, 1)`

> 这道题是**机器学习数据预处理**的核心步骤，非常实用。


````python
# 先自己写
````

## 参考解法


````python
data = np.array([[10., 20., 30.],
                 [2., 4., 6.],
                 [100., 200., 300.],
                 [5., 5., 5.]])
print('原始数据：')
print(data)
````

### 第 1 步：算每行的均值和标准差


````python
row_mean = data.mean(axis=1)
row_std = data.std(axis=1)

print('每行均值：', row_mean)
print('每行标准差：', row_std)
print('  <- 注意最后一行标准差是 0（三个数一样），马上会出问题')
````

### ⚠️ 陷阱：标准差为 0 会除以 0

第 4 行是 `[5, 5, 5]`，标准差是 0，除下去会变成 `nan`（Not a Number）。
**这是真实项目里最常见的坑之一**，处理方式见下面。


````python
z_raw = (data - row_mean.reshape(-1, 1)) / row_std.reshape(-1, 1)
print('直接标准化（注意最后一行的 nan）：')
print(z_raw)
````

### 正确做法：把 0 标准差替换成 1


````python
safe_std = np.where(row_std == 0, 1, row_std)     # 为 0 的地方换成 1
print('处理后的标准差：', safe_std)

z = (data - row_mean.reshape(-1, 1)) / safe_std.reshape(-1, 1)
print('\n标准化结果：')
print(z)

print('\n验证：每行均值 ->', np.round(z.mean(axis=1), 10))
print('验证：每行标准差 ->', np.round(z.std(axis=1), 6))
````

### 为什么必须 `reshape(-1, 1)`？

`row_mean` 形状是 `(4,)`，`data` 形状是 `(4, 3)`。
从右往左比：`4 vs 3` **不相等也没有 1，会报错或广播错**。

`reshape(-1, 1)` 把它变成 `(4, 1)`，再比：

```
data    : (4, 3)
row_mean: (4, 1)     <- 从右往左：3 vs 1 ✅（广播），4 vs 4 ✅
结果    : (4, 3)     每一行减去自己那一个数
```

**`reshape(-1, 1)` 是"把一维数组竖起来变成列向量"的标准写法，请牢记。**

# 第 9 题：处理缺失值（NaN）

**题目**：数组里混进了几个 `nan`，要求：

1. 找出 `nan` 的位置
2. 用**非缺失值**的平均数填充 `nan`

**知识点**：`np.isnan` / `np.nanmean` / `np.nan_to_num`

> ⚠️ 常见的错误想法：`nan` 用 `==` 判断不出来（`nan == nan` 是 `False`！），
> 必须用 `np.isnan`。


````python
# 先自己写
````

## 参考解法


````python
arr = np.array([1.0, 2.0, np.nan, 4.0, 5.0, np.nan, 7.0])
print('原始数组：', arr)
````

### 第 1 问：找出 nan 的位置

**先看一个反直觉的事实**：


````python
print('nan == nan 的结果：', np.nan == np.nan, '  <- False！')
print('所以不能用 == 来找 nan')
print('\n正确方式（np.isnan）：', np.isnan(arr))
print('nan 的下标：', np.where(np.isnan(arr))[0])
````

### 第 2 问：用平均值填充

**注意**：不能用 `np.mean(arr)`，它会被 nan "污染"成 nan。
要用 `np.nanmean`（会**忽略** nan 再求平均）。


````python
print('np.mean(arr)    =', np.mean(arr), '  <- 被污染了')
print('np.nanmean(arr) =', np.nanmean(arr), '  <- 忽略 nan，正确')
````

````python
filled = arr.copy()                       # 别改原数组，先复制一份
mask = np.isnan(filled)
filled[mask] = np.nanmean(filled)
print('填充后：', filled)
print('还有 nan 吗？', np.any(np.isnan(filled)))
````

### 顺带认识：`nan*` 系列全家桶

| 函数 | 作用 |
|---|---|
| `np.nansum` | 求和时忽略 nan |
| `np.nanmean` | 求平均时忽略 nan |
| `np.nanmax` / `np.nanmin` | 最值忽略 nan |
| `np.nanstd` / `np.nanvar` | 标准差/方差忽略 nan |

**凡是 `np.xxx` 前面加了个 `nan`，意思就是"算的时候把 nan 当空气"。**


````python
print('nansum :', np.nansum(arr))
print('nanmean:', np.nanmean(arr))
print('nanmax :', np.nanmax(arr))
````

# 第 10 题：排序与取 Top N

**题目**：给定 6 个学生的姓名和成绩，找出**成绩最高的 3 个人**（按从高到低）。

**知识点**：`argsort` / 花式索引 / 切片

> 这道题也解释了为什么 `argsort` 比 `sort` 更有用 ——
> 排序之后你还需要"谁是谁"。


````python
# 先自己写
````

## 参考解法


````python
names = np.array(['小明', '小红', '小刚', '小美', '小强', '小芳'])
scores = np.array([78, 92, 65, 88, 95, 71])

print('全部成绩：')
for n, s in zip(names, scores):
    print(f'  {n}: {s}')
````

### 关键：`argsort` 返回的是"下标顺序"，不是值


````python
order = np.argsort(scores)
print('argsort 结果（从小到大）：', order)
print('  <- 意思是：最低分在位置 2，然后 5、0、3、1、4')

print('\n用这个顺序取出【名字】：', names[order])
print('对应【成绩】：           ', scores[order])
````

### 从高到低 + 取前 3


````python
order_desc = np.argsort(scores)[::-1]         # [::-1] 翻转成从大到小
print('从高到低的下标：', order_desc)

top3 = order_desc[:3]                          # 取前 3 个下标
print('\n=== 成绩最高的 3 位 ===')
for i in top3:
    print(f'  {names[i]}：{scores[i]} 分')
````

### 换一种写法：把名字和成绩打包排序

工作中更常见的需求是"按分数排序输出完整名单"：


````python
print('完整排名：')
for rank, i in enumerate(order_desc, start=1):
    print(f'  第 {rank} 名：{names[i]}  {scores[i]} 分')
````

## 附加挑战（给学有余力的你）

**挑战 1**：不用循环，算出 `[1, 2, 3, 4, 5]` 的"相邻两数之差" `[1, 1, 1, 1]`
（提示：`np.diff`）。

**挑战 2**：给一个 `(10, 3)` 的随机矩阵，找出"每一行最大值所在的列号"
（提示：`argmax(axis=1)`）。

**挑战 3**：用 Numpy 实现"移动平均"：给 `[1,2,3,4,5,6]`，
计算窗口为 3 的移动平均 `[2, 3, 4, 5]`（提示：切片相加后除以 3）。

**挑战 4**：生成 10000 个正态分布随机数，统计落在
`[均值-1σ, 均值+1σ]` 范围内的比例（理论值约 68%），验证"正态分布的经验法则"。

---

**挑战题思路**

```python
# 1
print(np.diff([1, 2, 3, 4, 5]))

# 2
m = np.random.rand(10, 3)
print(m.argmax(axis=1))

# 3
a = np.array([1, 2, 3, 4, 5, 6], dtype=float)
print((a[:-2] + a[1:-1] + a[2:]) / 3)

# 4
x = np.random.normal(0, 1, 10000)
within = np.sum(np.abs(x) <= 1) / x.size
print(f'落在 ±1σ 内：{within:.1%}  （理论约 68.3%）')
```
