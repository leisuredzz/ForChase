## 一、数学运算函数

这类函数都是**逐个元素**地算，返回形状相同的新数组。


````python
print('np.sqrt 平方根      :', np.sqrt(arr))
print('np.exp  自然指数 e^x :', np.exp(arr))
print('np.log  自然对数 ln  :', np.log(arr))
print('np.abs  绝对值       :', np.abs(arr))
print('np.power 幂         :', np.power(arr, 2), '  <- 等价于 arr ** 2')
````

> **原稿错在哪**：原稿把 `np.exp([1,4,9,16])` 写成 `[2.718, 5.459, 81.03, 256]`。

> `e⁴ ≈ 54.6`、`e⁹ ≈ 8103`、`e¹⁶ ≈ 8.9×10⁶`。指数增长飞快，注释里那几个数根本对不上。


````python
print('逐项核对：')
print('  e^1 =', np.exp(1))
print('  e^4 =', np.exp(4))
print('  e^9 =', np.exp(9))
print('  e^16 =', np.exp(16))
````

## 二、统计计算函数

这类函数把数组"压"成一个数（或一组数）。


````python
print(f'sum      求和      : {np.sum(arr)}')
print(f'mean     平均值    : {np.mean(arr)}')
print(f'max      最大值    : {np.max(arr)}')
print(f'min      最小值    : {np.min(arr)}')
print(f'prod     连乘      : {np.prod(arr)}')
print(f'median   中位数    : {np.median(arr)}')
print(f'var      方差      : {np.var(arr)}')
print(f'std      标准差    : {np.std(arr)}')
````

### ⚠️ 这里原稿错了 5 处，逐个对账

让我们**手算验证**一遍 `arr = [1, 4, 9, 16]`：

| 指标 | 手算 | 手算结果 | 原稿写的 | 谁对 |
|---|---|---|---|---|
| `mean` | (1+4+9+16)/4 | **7.5** | 8.0 | ✅ 本篇对 |
| `prod` | 1×4×9×16 | **576** | 144 | ✅ 本篇对 |
| `var` | 均值 7.5，偏差平方和 129，129/4 | **32.25** | 24.0 | ✅ 本篇对 |
| `std` | √32.25 | **5.6789** | 4.8989 | ✅ 本篇对 |
| `median` | 中间两数 (4+9)/2 | **6.5** | 8.0 | ✅ 本篇对 |

**原稿大概是照抄了另一组数据的输出**（那组数据看起来均值是 8、中位数 8）。
这提醒我们：**注释里的输出一定要自己跑一遍**，抄来的最不靠谱。


````python
# 用代码把"手算"重做一遍，让它自我验证
m = arr.mean()
print('手算 mean   :', arr.sum() / arr.size)
print('手算 prod   :', 1 * 4 * 9 * 16)
print('手算 var    :', ((arr - m) ** 2).sum() / arr.size)
print('手算 std    :', np.sqrt(((arr - m) ** 2).sum() / arr.size))
print('手算 median :', np.median(arr), '（中间两个数 4 和 9 的平均）')
````

### 累计型：不是压成一个数，而是"边走边算"


````python
print('np.cumsum  累计和  :', np.cumsum(arr), '  <- [1, 1+4, 1+4+9, 1+4+9+16]')
print('np.cumprod 累计乘  :', np.cumprod(arr), '  <- [1, 1×4, 1×4×9, ...]')
````

### 分位数 / 百分位数（数据分析和画箱线图的重点）

- `np.percentile(a, 25)` —— 参数是 **0~100 的百分数**
- `np.quantile(a, 0.25)` —— 参数是 **0~1 的小数**

两者是一回事，只是刻度不同：`percentile(a, 25)` == `quantile(a, 0.25)`。


````python
print('percentile :', np.percentile(arr, 25), np.percentile(arr, 50), np.percentile(arr, 75))
print('quantile   :', np.quantile(arr, 0.25), np.quantile(arr, 0.50), np.quantile(arr, 0.75))
print('两者相等吗？', np.array_equal(
    [np.percentile(arr, 25), np.percentile(arr, 50), np.percentile(arr, 75)],
    [np.quantile(arr, 0.25), np.quantile(arr, 0.50), np.quantile(arr, 0.75)],
))
````

> **原稿错在哪**：原稿写 25% / 50% / 75% 分位数是 `3.75 / 8.0 / 12.25`。
> 实际是 **`3.25 / 6.5 / 10.75`**。
> 而且注意 `50%` 分位数**就是中位数**，应该等于上面算的 `6.5`，原稿自己前后矛盾了。

**分位数是怎么插出来的？**（Numpy 默认 `linear` 方法）

```
位置 = q × (n - 1)      q=0.25, n=4  ->  位置 = 0.75
值   = a[0] + 0.75 × (a[1] - a[0])
     = 1    + 0.75 × (4 - 1)
     = 3.25
```


````python
print('手算 25% 分位：位置 =', 0.25 * (4 - 1), '，值 =', arr[0] + 0.75 * (arr[1] - arr[0]))
````

## 三、比较与逻辑函数

这类函数返回**布尔数组**，是数据筛选的基础。


````python
a1 = np.array([1, 2, 3])
a2 = np.array([1, 2, 4])

print('np.equal       :', np.equal(a1, a2), '  <- 等价于 a1 == a2')
print('np.not_equal   :', np.not_equal(a1, a2))
print('np.greater     :', np.greater(a1, a2), '  <- 等价于 a1 > a2')
print('np.greater_equal:', np.greater_equal(a1, a2))
print('np.less        :', np.less(a1, a2))
print('np.less_equal  :', np.less_equal(a1, a2))
````

### `np.all` 和 `np.any`：把布尔数组"压"成一个结论


````python
bool_arr = np.array([True, True, False])
print('数组           :', bool_arr)
print('np.all（全为真？）:', np.all(bool_arr), '  <- 有一个 False 就是 False')
print('np.any（有真吗？）:', np.any(bool_arr), '  <- 有一个 True 就是 True')

# 常见用法：判断"是否全部满足条件"
print('\narr 是否全部大于 0 ？', np.all(arr > 0))
print('arr 是否存在大于 10 的？', np.any(arr > 10))
````

> **原稿问题**：原稿里 `np.all()` 和 `np.any()` **各出现了两遍**（编号 1、2 和 12、13 重复）。
> 这里合并去重，顺便补上最实用的"配合条件表达式"用法。


### 逻辑运算：`logical_and` / `logical_or` / `logical_not`


````python
print('np.logical_and :', np.logical_and(a1, a2), '  <- 逐元素 与（非 0 即真）')
print('np.logical_or  :', np.logical_or(a1, a2))
print('np.logical_not :', np.logical_not(a1), '  <- 非 0 全变 False')
````

**实际写代码时，更常用 `&` `|` `~` 这三个符号**，但它们有优先级陷阱：


````python
arr5 = np.array([1, 4, 9, 16])

# 想筛"大于 3 且小于 10"的元素
print('正确（加括号）：', arr5[(arr5 > 3) & (arr5 < 10)])

# 不加括号会报错或语义错乱，因为 & 的优先级比 > 高
# print(arr5[arr5 > 3 & arr5 < 10])     # 会报错
````

> **铁律**：用 `&` `|` `~` 组合多个条件时，**每个条件都套一对圆括号**。
> （`and` / `or` / `not` 关键字在 Numpy 数组上**不能用**，它们只处理单个布尔值。）

## 四、条件选择：`np.where` 和 `np.select`

### `np.where(条件, 真值, 假值)` —— 三目运算符的数组版


````python
arr3 = np.array([10, 20, 30])
print('np.where(arr3 > 20, arr3, 0) :', np.where(arr3 > 20, arr3, 0), '  <- 大于 20 保留，否则填 0')

# 二值化：大于 2 记 1，否则记 0
arr4 = np.array([1, 2, 3, 4, 5])
print('np.where(arr4 > 2, 1, 0)     :', np.where(arr4 > 2, 1, 0))

# 只有一个参数时（不写真假值），返回"满足条件的下标"
print('\n只有一个参数：np.where(arr3 > 20) ->', np.where(arr3 > 20), '  <- 返回下标组成的元组')
````

### 多级分类：用 `np.where` 嵌套（不推荐）vs `np.select`（推荐）

原稿这里有一段**语法就是错的**代码（把参数写成了一个元组）：

```python
np.where((score<=60, '不及格', np.where(...)))    # ✗ 这是错的
```

下面给出**两种能跑的写法**，对比一下你就知道该用哪个。


````python
np.random.seed(0)
score = np.random.randint(0, 100, 10)
print('成绩：', score)
````

**写法 A：`np.where` 嵌套** —— 能跑，但嵌套一多就看不懂了


````python
grade_a = np.where(score >= 90, '优秀',
                   np.where(score >= 60, '及格', '不及格'))
print('写法 A（嵌套 where）：', grade_a)
````

**写法 B：`np.select`** —— 条件自上而下匹配，可读性强得多


````python
grade_b = np.select(
    [score >= 90, score >= 60],        # 条件列表，注意顺序：先判断严格的
    ['优秀', '及格'],                   # 对应结果
    default='不及格',                   # 都不满足时的兜底
)
print('写法 B（select）    ：', grade_b)
print('\n两种写法结果一致吗？', np.array_equal(grade_a, grade_b))
````

> **记住顺序陷阱**：`np.select` 是**从上往下第一个匹配的条件生效**，
> 所以 `>= 90` 必须写在 `>= 60` 前面。
> 如果写反了，`>= 60` 会先"吃掉"所有 60 分以上的人，优秀就永远选不出来。

## 五、排序与去重


````python
np.random.seed(0)

arr_sort = np.array([3, 1, 4, 1, 5, 9, 2, 6, 5])
print('原数组        :', arr_sort)
print('np.sort       :', np.sort(arr_sort), '  <- 返回排好序的【新数组】')
print('原数组还在吗？ :', arr_sort, '  <- 在，没被改')

# 想让原数组本身被排序，用 .sort() 方法（注意：没括号返回值，是原地改）
arr_sort.sort()
print('调用 .sort() 后 :', arr_sort)
````

### `np.sort` vs `arr.sort()` —— 这是个大坑

| 写法 | 是否改原数组 | 返回值 |
|---|---|---|
| `np.sort(arr)` | **不改** | 排好序的新数组 |
| `arr.sort()` | **原地改** | `None` |

所以 `arr = arr.sort()` 是**经典错误** —— 会把 arr 变成 `None`。


````python
print('np.argsort 返回排序后的下标 :', np.argsort(np.array([3, 1, 4, 1, 5, 9, 2, 6, 5])))
print('  <- 意思是：最小的在 idx 1，第二小在 idx 3，第三小在 idx 6 ...')

# 用处：按某列排序另一个数组
values = np.array([88, 92, 75])
names = np.array(['小明', '小红', '小刚'])
order = np.argsort(values)[::-1]         # 从大到小
print('\n按分数从高到低：', names[order], values[order])
````

### 去重：`np.unique`


````python
dup = np.array([1, 2, 2, 3, 4, 4, 5])
print('原数组      :', dup)
print('np.unique   :', np.unique(dup), '  <- 去重并【自动排序】')

# return_counts=True 能顺便统计每个值出现的次数
uniq, counts = np.unique(dup, return_counts=True)
print('\n唯一值 :', uniq)
print('出现次数:', counts)

for v, c in zip(uniq, counts):
    print(f'  {v} 出现了 {c} 次')
````

> **小提示**：`np.unique` 返回的**永远是排好序的**，
> 所以它不能用来"保持原顺序去重"。要保持原顺序得另想办法（比如 pandas 的 `drop_duplicates`）。


### `np.partition`：只排出"最小的 k 个"，不做全排序


````python
p = np.array([3, 1, 4, 1, 5, 9, 2, 6, 5])
print('原数组           :', p)
print('np.partition(p, 3):', np.partition(p, 3))
print('  <- 保证【位置 0~2】是全场最小的 3 个（内部顺序不保证），位置 3 是第 4 小的')

# 验证：前 3 个确实是最小的 3 个
print('前 3 个排序后：', np.sort(np.partition(p, 3)[:3]))
print('全场最小的3个：', np.sort(p)[:3])
````

**为什么要用 `partition`？** 因为"只要最小的 3 个"时，
全排序 `O(n log n)` 是浪费，`partition` 只要 `O(n)`。数据量大时差别明显。

## 六、变形与拼接


````python
# reshape：改变形状（元素总数必须对得上）
r = np.arange(6)
print('np.reshape(arr, (2,3)) :')
print(np.reshape(r, (2, 3)))
print('  <- 一共 6 个元素，改成 2 行 3 列')

# -1 让 Numpy 自己算那一维
print('\n用 -1 自动推断：(2, -1) ->')
print(r.reshape(2, -1))

# 注意：reshape 通常是"视图"，改一个会影响另一个
print('\nreshape 返回的是视图吗？', r.reshape(2, 3).base is r)
````

````python
# concatenate：沿某个轴拼接
arr_x = np.array([1, 2, 3])
arr_y = np.array([4, 5, 6])
print('np.concatenate 一维 :', np.concatenate((arr_x, arr_y)))
print('注意和 arr_x + arr_y 的区别 :', arr_x + arr_y, '  <- 这是逐项相加，不是拼接！')

# 二维沿不同轴拼接
m1 = np.array([[1, 2], [3, 4]])
m2 = np.array([[5, 6], [7, 8]])
print('\naxis=0（上下叠）：')
print(np.concatenate((m1, m2), axis=0))
print('axis=1（左右拼）：')
print(np.concatenate((m1, m2), axis=1))
````

````python
# hstack / vstack：更直观的简写
print('np.hstack 水平拼 :')
print(np.hstack((m1, m2)))
print('\nnp.vstack 垂直叠 :')
print(np.vstack((m1, m2)))
````

````python
# split：切分
s = np.array([1, 2, 3, 4, 5, 6])
print('np.split(s, 3)  :', np.split(s, 3), '  <- 平均切成 3 份')
print('\nnp.split 返回的是列表：', type(np.split(s, 3)))
````

## 七、常见坑清单

| 坑 | 现象 | 正解 |
|---|---|---|
| 注释里的输出是从别处抄的 | 和实际对不上 | **每个输出都真跑一遍** |
| `arr.sort()` 的返回值 | 是 `None` | 原地排序不返回值；要新数组用 `np.sort(arr)` |
| 多条件用 `and`/`or` | `ValueError` | 用 `&` `|` `~`，且每个条件加括号 |
| `np.where` 嵌套三层以上 | 看不懂 | 改用 `np.select` |
| `np.select` 条件顺序 | 结果不符合预期 | **严格的条件写在前面** |
| `percentile` / `quantile` 参数混用 | 结果差 100 倍 | 前者 0~100，后者 0~1 |
| `np.unique` 保持原顺序去重 | 做不到 | 它一定排序；保持顺序用 pandas |
| `reshape` 元素数对不上 | `ValueError` | 总元素数必须相等，可用 `-1` 自动推断 |
| `np.concatenate` 和 `+` 混用 | 一个是拼接一个是相加 | 记清语义 |
| 方差跨库对比 | numpy ≠ pandas | ddof 默认不同（0 vs 1） |
