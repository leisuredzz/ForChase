## 一、算术运算符（7 个）

| 运算符 | 名称 | 例子 | 结果 | 记忆点 |
|---|---|---|---|---|
| `+` | 加 | `5 + 2` | `7` | |
| `-` | 减 | `5 - 2` | `3` | |
| `*` | 乘 | `5 * 2` | `10` | |
| `/` | 除 | `5 / 2` | `2.5` | **结果永远是浮点数** |
| `//` | 整除（向下取整） | `5 // 2` | `2` | 丢掉小数部分 |
| `%` | 取余 | `5 % 2` | `1` | 只要余数 |
| `**` | 幂 | `5 ** 2` | `25` | 5 的 2 次方 |


````python
print('加法：', 1 + 2)
print('减法：', 1 - 2)
print('乘法：', 1 * 2)
print('除法：', 1 / 2)          # 0.5 —— 注意：即使能整除，结果也是浮点
print('8 / 2 =', 8 / 2)          # 4.0，不是 4！
print('整除：', 9 // 2)          # 4
print('取余：', 9 % 2)           # 1
print('幂：', 9 ** 2)            # 81
````

### 1.1 `/` vs `//`：最容易搞混的一对

```python
9 / 2   ->  4.5    普通除法，结果一定是 float
9 // 2  ->  4      整除，目的是"只要整数部分"
```

**但 `//` 不是简单的"去掉小数"，它是"向下取整"（往数轴的负方向靠）。**
这一点在负数上会让人意外：


````python
print('-9 // 2 =', -9 // 2)     # -5，不是 -4！
print('-9 /  2 =', -9 / 2)      # -4.5，向下取整就是 -5
print('7 // 2  =', 7 // 2)      # 3
print('round(-4.5) 是四舍五入：', round(-4.5))   # -4（银行家舍入，和 // 不同）
````

### 1.2 取余的实用场景（这个真的很常用）

- 判断奇偶：`x % 2 == 0` 是偶数
- 判断整除：`x % 3 == 0` 能被 3 整除
- 循环轮转：`第 i 个 % 3` 可以得到 `0,1,2,0,1,2...` 的循环下标
- 求时间：秒数转"几分几秒"


````python
# 场景 1：奇偶判断
for n in range(1, 6):
    print(f'{n} 是{"偶" if n % 2 == 0 else "奇"}数', end='  ')
print()

# 场景 2：秒数转分秒
total_seconds = 3725
print(f'{total_seconds} 秒 = {total_seconds // 60} 分 {total_seconds % 60} 秒')

# 场景 3：三列轮转（把一列数据分到三组去）
for i in range(8):
    group = i % 3
    print(f'第 {i} 个 -> 第 {group} 组')
````

### 1.3 整除和取余天生是一对

公式：`被除数 = 除数 × 商 + 余数`

```python
a = 13, b = 3
a // b = 4   (商)
a %  b = 1   (余)
验证：3 * 4 + 1 = 13  ✓
```

Python 内置了一个 `divmod()`，一次同时拿到商和余数：


````python
a, b = 13, 3
print('divmod(13, 3) =', divmod(a, b))       # (4, 1) 一个元组
print('pow(2, 10) =', pow(2, 10))            # 1024，等价于 2 ** 10
````

## 二、字符串也能"算"

两个运算符对字符串有特殊含义：


````python
print('-' * 50)                          # * 是重复：把字符串重复 50 次（画分隔线神器）
print('Ha' * 3)                          # HaHaHa
print('Hello' + ' ' + 'World')           # + 是拼接：首尾相接

# 列表也支持 + 和 *
print([1, 2, 3] + [4, 5, 6])             # 拼接
print([0] * 5)                           # [0, 0, 0, 0, 0]

# 【坑】+ 号两边类型必须一致
try:
    print('年龄：' + 18)
except TypeError as e:
    print('报错：', e)
    print('正解 1：', '年龄：' + str(18))
    print('正解 2：', f'年龄：{18}')      # 推荐用 f-string

# 列表 + 列表是拼接，而不是"对应元素相加"：
print('[1,2,3] + [4,5,6] =', [1, 2, 3] + [4, 5, 6])
# 想对应元素相加，要用循环、推导式，或者 numpy（数据分析那章会讲）：
print('对应元素相加：', [x + y for x, y in zip([1, 2, 3], [4, 5, 6])])
````

## 三、赋值运算符：写起来更短，读起来更快

| 简写 | 等价于 | 例子（a 初始为 10） |
|---|---|---|
| `a += b` | `a = a + b` | `a += 3` → 13 |
| `a -= b` | `a = a - b` | `a -= 3` → 7 |
| `a *= b` | `a = a * b` | `a *= 3` → 30 |
| `a /= b` | `a = a / b` | `a /= 4` → 2.5 |
| `a //= b` | `a = a // b` | `a //= 3` → 3 |
| `a %= b` | `a = a % b` | `a %= 3` → 1 |
| `a **= b` | `a = a ** b` | `a **= 2` → 100 |


````python
x = 10
x = x + 3        # 啰嗦写法
print('x = x + 3 ->', x)
x += 3           # 简写：结果一样，但少写一遍 x
print('x += 3    ->', x)

a = 10
b = 20
a *= b
print('a *= b ->', a)

# 【注意】+= 对列表来说不是"复制后再拼接"，而是【原地修改】！
list_a = [1, 2]
list_b = list_a
list_a += [3]           # 等价于 list_a.extend([3])，原地改
print('list_a =', list_a, ' list_b =', list_b)   # 两个都变了

list_c = [1, 2]
list_d = list_c
list_c = list_c + [3]   # 这次是【新建】一个列表再赋值
print('list_c =', list_c, ' list_d =', list_d)   # list_d 没变
# 记法：+= 会改原对象，+ 会造新对象。这个区别在函数传参时特别重要。
````

## 四、多变量赋值与解包（Python 的招牌语法）

### 4.1 同时赋值


````python
a, b, c = 1, 2, 3
print(a, b, c)          # 1 2 3

# 交换两个变量，不需要临时变量 —— 这是 Python 独有的优雅写法
a, b = b, a
print('交换后：', a, b)
````

### 4.2 星号 `*` 收集多余的值

规则：**带 `*` 的变量只能有一个，它会收集"剩下的所有值"成为一个列表。**


````python
a, *b, c = 1, 2, 3, 4, 5
print('a =', a)      # 1
print('b =', b)      # [2, 3, 4]  —— 中间的都归它
print('c =', c)      # 5

*a, b = 1, 2, 3, 4, 5
print('a =', a, ' b =', b)      # [1, 2, 3, 4]  5

first, second, *rest = [1, 2, 3, 4, 5, 6]
print('first =', first, ' second =', second, ' rest =', rest)

# 【坑】星号变量不能放中间只有一个位置时会报错，但"中间"本身是可以的：
#   a, *b, *c = ...   -> SyntaxError: 只能有一个星号
# 星号变量也可以是空列表：
x, *y = [1]
print('x =', x, ' y =', y)      # 1 []
````

## 五、比较运算符与逻辑运算符

### 5.1 比较运算符：结果永远是 `True` / `False`


````python
print(5 > 3, 5 < 3, 5 >= 5, 5 <= 3, 5 == 5, 5 != 3)

# Python 支持"链式比较"，写数学不等式一样自然
score = 78
print('60 <= score < 80 :', 60 <= score < 80)      # True
# 等价于 60 <= score and score < 80，但只算一次 score，更简洁也更快

# 【重点坑】== 和 is 不是一回事
a = [1, 2, 3]
b = [1, 2, 3]
print('a == b :', a == b)     # True  —— 值相同
print('a is b :', a is b)     # False —— 不是同一个对象
print('a is a :', a is a)     # True

x = 256
y = 256
print('小整数 256 is 256 :', x is y)    # True（Python 缓存了小整数 -5~256）
````

**结论**：判断"值是否相等"永远用 `==`；`is` 只在判断 `None`、`True`、`False` 时用（`if x is None:`）。


### 5.2 逻辑运算符：`and` / `or` / `not`

| 运算符 | 含义 | 记忆 |
|---|---|---|
| `and` | 两边都真才为真 | 见好就收：遇到假立刻返回 |
| `or` | 只要一边真就是真 | 见好就收：遇到真立刻返回 |
| `not` | 取反 | `not True` → `False` |


````python
high_income = True
good_credit = True
student = True

# 贷款审批：收入高或信用好，且不是学生
if (high_income or good_credit) and not student:
    print('eligible')
else:
    print('not eligible')
````

### 5.3 短路求值：`and`/`or` 会"提前收工"

这是逻辑运算符最重要的特性，也是很多代码简洁写法的来源：

- `A and B`：**A 为假，直接返回 A**，B 根本不会被执行
- `A or B`：**A 为真，直接返回 A**，B 根本不会被执行

利用这一点可以：**先判断"能不能安全操作"，再做操作**。


````python
# 场景 1：避免除零错误 —— 用 and 短路来保护
data = []
if data and data[0] != 0 and 100 / data[0] > 5:
    print('数据正常')
else:
    print('data 是空的，第一个条件就短路了，没有触碰 data[0]，不会报错')

# 场景 2：or 做默认值（很常用的写法）
name = None
print('显示名：', name or '匿名用户')     # name 是 None（假值），所以返回右边

# 场景 3：or 的返回值其实是"操作数本身"，不一定是布尔值
print(0 or 'fallback ->', 0 or 'fallback')
print('abc' or 'fallback ->', 'abc' or 'fallback')   # 左边为真，直接返回 abc
````

`and` / `or` 返回的是**操作数本身**，这点很多人第一次看到会懵：


````python
print("0 or  '默认值'  ->", 0 or '默认值')       # 默认值
print("'' or '默认值'  ->", '' or '默认值')      # 默认值
print("'有值' or '默认值' ->", '有值' or '默认值')  # 有值
print("1 and 2 ->", 1 and 2)                     # 2
print("0 and 2 ->", 0 and 2)                     # 0

# 实战：给函数参数兜底
def greet(name=None):
    name = name or '陌生人'      # 简洁的默认值处理
    return f'你好，{name}'

print(greet())
print(greet('Cedric'))
````

### 5.4 成员运算符：`in` / `not in`


````python
fruits = ['apple', 'banana', 'cherry']
print("'banana' in fruits  ->", 'banana' in fruits)
print("'grape' not in fruits ->", 'grape' not in fruits)
print("'app' in 'apple'    ->", 'app' in 'apple')       # 也可以用于字符串（子串判断）
print("3 in {1, 2, 3}      ->", 3 in {1, 2, 3})
print("'name' in {'name': 1} ->", 'name' in {'name': 1})  # 字典查的是【键】
````

## 六、位运算符（了解即可，做算法和底层优化时才用）

| 运算符 | 名称 | 例子 |
|---|---|---|
| `&` | 按位与 | `6 & 3` → 2 |
| `\|` | 按位或 | `6 \| 3` → 7 |
| `^` | 按位异或 | `6 ^ 3` → 5 |
| `~` | 按位取反 | `~6` → -7 |
| `<<` | 左移 | `1 << 3` → 8（等价于 1×2³） |
| `>>` | 右移 | `8 >> 3` → 1（等价于 8÷2³） |


````python
print('6 & 3  =', 6 & 3)        # 110 & 011 = 010 = 2
print('6 | 3  =', 6 | 3)        # 110 | 011 = 111 = 7
print('6 ^ 3  =', 6 ^ 3)        # 110 ^ 011 = 101 = 5
print('1 << 3 =', 1 << 3)       # 8
print('8 >> 3 =', 8 >> 3)       # 1

# 实用场景：用位运算做"标志位"判断权限
READ, WRITE, EXEC = 1, 2, 4     # 分别是二进制的 001, 010, 100
perm = READ | WRITE             # 读+写
print('能读吗？', bool(perm & READ))
print('能写吗？', bool(perm & WRITE))
print('能执行吗？', bool(perm & EXEC))
````

## 七、运算符优先级（记不住就把括号写上）

从高到低（常用部分）：

```
1.  ()                 括号
2.  **                 幂
3.  * / // %           乘除类
4.  + -                加减
5.  << >>              移位
6.  &                  按位与
7.  ^ |                按位异或 / 或
8.  == != > >= < <=    比较
9.  not                逻辑非
10. and                逻辑与
11. or                 逻辑或
```

**一条实用建议**：与其背优先级，不如**多加括号**。
括号不花钱，但能让读代码的人（包括三个月后的你）少想两秒钟。


````python
print(2 + 3 * 4)         # 14，先乘后加
print((2 + 3) * 4)       # 20，括号优先
print(2 ** 3 ** 2)       # 512，幂运算【从右往左】结合：2**(3**2) = 2**9
print(not 1 == 2 or True and False)   # 括号化理解：(not (1 == 2)) or (True and False) = True
````

## 八、常见坑清单

| 坑 | 说明 | 正解 |
|---|---|---|
| `7 / 2` 以为是 3 | `/` 结果永远是 float | 要整数用 `7 // 2` |
| `-9 // 2` 以为是 -4 | `//` 是向下取整 | 得到 -5；要"截断"用 `int(-9 / 2)` |
| `'a' + 1` | 类型不匹配 | `'a' + str(1)` 或 f-string |
| `a is b` 当 `==` 用 | `is` 比的是身份不是值 | 值比较用 `==` |
| `[1,2] + [3,4]` 以为元素相加 | `+` 是列表拼接 | 用 `zip` + 推导式 |
| `list_a += [x]` 后 list_b 也变了 | `+=` 是原地修改 | 要独立副本用 `+` 或 `.copy()` |
| `a, *b, *c = ...` | 只能有一个星号 | 改成 `a, *b, c` |
| `1 and 2` 以为返回 True | `and`/`or` 返回操作数本身 | 需要布尔值就套 `bool()` |
| `0 or x` 当判断用 | 0 是假值，会返回 x | 判断存在性用 `is None` |
