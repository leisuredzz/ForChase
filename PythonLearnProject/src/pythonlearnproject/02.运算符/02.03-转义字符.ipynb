# 转义字符：在字符串里"插入看不见的东西"

---

**学习目标**

1. 理解"转义"这两个字到底在干什么
2. 掌握常用的 8 个转义字符
3. 知道什么时候该用原始字符串 `r""`
4. 明白为什么用 `\t` 对齐中文表格一定会翻车

**一句话理解**

转义就是"给特殊符号开一个后门"。
反斜杠 `\` 相当于告诉 Python："接下来这个字符你别按常规理解，按我说的特殊含义来。"


---
## 一、什么是"转义"

在字符串里，有些字符**没法直接打出来**（比如换行），
有些字符**本身有语法含义**（比如引号）。

反斜杠 `\` 就是解决这两个问题的钥匙：**`\` + 一个字符 = 一个新含义**。

```
\n  ->  换行符（看起来是两个字，其实是一个字符）
\'  ->  单引号本身（而不是"字符串结束"的信号）
```


````python
# 验证一下："\n" 是两个字符还是一个？
s = '\n'
print('len("\\n") =', len(s))          # 1 —— 确实只有一个字符
print('它到底是什么码点：', ord(s))     # 10，ASCII 里的换行符
````

---
## 二、8 个必会转义字符

| 写法 | 名称 | 作用 |
|---|---|---|
| `\n` | 换行符 | 光标移到下一行行首 |
| `\t` | 制表符 | 相当于一个 Tab 的距离 |
| `\\` | 反斜杠 | 输出一个真正的 `\` |
| `\'` | 单引号 | 在单引号字符串里输出 `'` |
| `\"` | 双引号 | 在双引号字符串里输出 `"` |
| `\r` | 回车 | 光标回到本行行首（会覆盖前面的内容） |
| `\b` | 退格 | 删除前一个字符 |
| `\0` | 空字符 | C 语言风格的字符串结束符，Python 里很少用 |

另外还有两个"按编码输出"的写法：

| 写法 | 作用 | 例子 |
|---|---|---|
| `\uXXXX` | 按 Unicode 码点输出 | `\u4f60\u597d` → `你好` |
| `\xXX` | 按十六进制字节输出 | `\x41` → `A` |


````python
# 1) \n 换行
print('第一行\n第二行')
print('用三行代码画一个三角形：')
print('  *\n ***\n*****')
````

````python
# 2) \t 制表符
print('列1\t列2\t列3')
print('A\tB\tC')
````

````python
# 3) \\ 反斜杠本身
print('这是一个反斜杠：\\')
print('Windows 路径若不转义会出问题：C:\\Users\\Cedric')
print('不转义会怎样？看看这个：')
print('C:\new_folder')     # \n 被当成换行了！
````

````python
# 4) 引号转义
print('这是一个单引号：\'')
print("这是一个双引号：\"")
print('we\'re friend')
````

### 关于引号：其实有更省事的办法

**外层用双引号，里层就能直接用单引号**，不需要转义：


````python
print("it's a nice day")      # 外层双引号，内层单引号，不用 \
print('他说"你好"')            # 外层单引号，内层双引号，也不用 \
print("it's a \"nice\" day")  # 同种引号才必须转义
````

````python
# 5) \r 回车：光标回到行首，后面的内容会覆盖前面
print('进度：0%\r进度：100%')     # 你会只看到 "进度：100%"

# 这在命令行里可以做"原地刷新"的进度条效果：
import time, sys
for i in range(0, 101, 25):
    sys.stdout.write(f'\r下载中 {"#" * (i // 5)}{"." * (20 - i // 5)} {i}%')
    sys.stdout.flush()
    time.sleep(0.1)
print()
````

````python
# 6) \b 退格：删掉前面一个字符
print('abc\bX')          # abX —— \b 把 c 删掉，然后写入 X
print('12345\b\b\bABC')  # 12ABC
````

````python
# 7) \uXXXX 按 Unicode 码点输出（记不住码点可以查 ord/chr）
print('\u4f60\u597d')                 # 你好
print('笑脸：\u2764')                  # 一个爱心符号
print('用 chr 反查：', chr(0x4f60))     # 你
````

## 三、原始字符串 `r""`：让转义全部失效

在字符串前面加一个 `r`，整个字符串里的 `\` 就**不再有特殊含义**，全部原样输出。


````python
print('hello\nhello')       # 真的换行了
print(r'hello\nhello')      # hello\nhello —— 原样打印
print(r'C:\Users\Cedric\new_folder')     # 路径终于不用一个个加反斜杠了
print(r'\d{3}-\d{8}')       # 正则表达式里写 \d 不用变成 \\d
````

**什么时候用 `r""`？**

- 写 Windows 路径
- 写正则表达式（比如 `re.search(r'\d+', s)`）
- 字符串里有大量反斜杠，懒得一个个转义

**注意**：`r""` 里不能以单个 `\` 结尾，因为那会把结尾的引号转义掉。

## 四、【修 bug 现场】为什么用 `\t` 对齐表格一定会翻车

原始笔记里有这么一段：

```python
str1 = '姓名\t\t\t年龄\t\t\t\t家庭住址\t\t\t\t\t状况'
str2 = '老马\t\t\t22\t\t\t\t河北邯郸\t\t\t\t\t恋爱中'
str3 = '老连\t\t\t22\t\t\t\t湖北孝感\t\t\t\t\t单身'
print(str1); print(str2); print(str3)
```

作者的意图是对齐成一张表，但**只要数据一变长度，又得重新数 Tab**，
而且中文字符宽度是英文的 2 倍，即使 Tab 数量一样也常常对不齐。

先看看实际效果有多歪：


````python
rows = [
    '姓名\t\t\t年龄\t\t\t\t家庭住址\t\t\t\t\t状况',
    '老马\t\t\t22\t\t\t\t河北邯郸\t\t\t\t\t恋爱中',
    '老连\t\t\t22\t\t\t\t湖北孝感\t\t\t\t\t单身',
    '欧阳夏侯\t\t\t22\t\t\t\t内蒙古自治区\t\t\t\t\t单身',
]
for r in rows:
    print(r)
````

**正解：不要用 `\t` 手工凑宽度，用 f-string 的格式规格来指定列宽。**

格式：`{值:<宽度}` 左对齐、`{值:>宽度}` 右对齐、`{值:^宽度}` 居中。


````python
def show_table(rows):
    # 表头也参与对齐，宽度写成常量，后面统一引用
    W_NAME, W_AGE, W_ADDR, W_STATUS = 12, 6, 16, 8
    header = f'{"姓名":<{W_NAME}}{"年龄":>{W_AGE}}{"家庭住址":<{W_ADDR}}{"状况":<{W_STATUS}}'
    print(header)
    print('-' * len(header))
    for name, age, addr, status in rows:
        print(f'{name:<{W_NAME}}{age:>{W_AGE}}{addr:<{W_ADDR}}{status:<{W_STATUS}}')


show_table([
    ('老马', 22, '河北邯郸', '恋爱中'),
    ('老连', 22, '湖北孝感', '单身'),
    ('欧阳夏侯', 22, '内蒙古自治区', '单身'),
])
````

这样就整齐多了，而且**新增行、改数据都不用重新数 Tab**。

> 唯一的小瑕疵：中文字符在终端里占 2 个字符宽，而 Python 按 1 个字符算长度，
> 所以纯中文的列之间仍然会有细微错位。要完全对齐需要按"显示宽度"计算，
> 可以用 `wcwidth` 库，或者干脆把中文列留给 Excel/HTML 去排版。

## 五、常见坑清单

| 坑 | 现象 | 正解 |
|---|---|---|
| Windows 路径没转义 | `'C:\new_folder'` 里的 `\n` 变成换行 | 用 `r'C:\new_folder'` |
| 字符串末尾单个 `\` | `SyntaxError: unterminated string literal` | 写成 `'\\'` |
| 用 `\t` 对齐中文表格 | 对不齐、加数据还要重数 | 用 f-string 的 `:<宽度` |
| 以为 `\n` 是 2 个字符 | `len('\n')` 是 1 | 转义后是**一个**字符 |
| `r""` 里以为还能用 `\n` 换行 | 原样输出 `\n` | 要换行就别加 `r` |
| 三引号字符串里用了 `\` 结尾 | 报错 | 检查每行结尾 |
| 打印反斜杠写 `print('\\')` 却打成 `print('\')` | 报错 | 反斜杠自己也要转义 |
